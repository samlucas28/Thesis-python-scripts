```python
# ----------------------------------------------------------------
# FULL MODEL
# ----------------------------------------------------------------
# 1. Defining the cohort
# 2. Distributions
# 3. Decision tree
  # 3.1 Inputs
  # 3.2 Node classes
  # 3.3 Building
  # 3.4 Simulate function
# 4. Markov model 1: Unclassified General population
  # 4.1 Model setup
  # 4.2 Costs
  # 4.3 Probabilities
  # 4.4 Transition probabilities
  # 4.5 Simulation function
# 5. Markov model 2: Moderate/high risk
  # 5.1 Model setup
  # 5.2 Cost parameters
  # 5.3 Load prevelance data
  # 5.4 Helper functions
  # 5.5 Matrix builder
  # 5.6 Cancer detection function
  # 5.7 Costs calculator
  # 5.8 Model 2 simulation function
# 6. Probabilistic sensitivity analysis (Full loop)
# 7. Deterministic analysis
  # 7.1 Define deterministic
  # 7.2 Simulate decision tree
  # 7.3 Simulate full model
  # 7.4 One-way sensitivity analysis
    # 7.4.1 Define simulation function
    # 7.4.2 Parameter ranges
    # 7.4.3 Summary table
    # 7.4.4 Tornado Plot
# 8. Distributional cost-effectiveness analyses
  # 8.1 FHQS uptake by imd
  # 8.2 Simulate full model for DCEA
  # 8.3 Outcome distributions pre and post intervention
  # 8.4 Inequality measures
  # 8.5 Equity weighted analysis
    # 8.5.1 Calculate and normalise weights
    # 8.5.2 Apply weights
    # 8.5.3 Display plot




from __future__ import annotations
from dataclasses import dataclass
from typing import Dict, List, Tuple, Callable, Union
import numpy as np
from scipy.stats import beta, gamma, dirichlet, lognorm
import pandas as pd

# See python version
import sys
print(sys.version)

# 1. DEFINING THE COHORT OF WOMEN
# -----------------------------------
import numpy as np
import pandas as pd

rng = np.random.default_rng(seed=1)

N = 50000

# Read Age × IMD distribution from Excel
age_imd = pd.read_excel(
    "/content/drive/My Drive/Colab Notebooks/UK age and IMD distribution.xlsx"
)

# Convert wide format to long format
age_imd_long = age_imd.melt(
    id_vars="IMD",
    var_name="start_age",
    value_name="population"
)

# Convert to numeric
age_imd_long["start_age"] = age_imd_long["start_age"].astype(int)
age_imd_long["IMD"] = age_imd_long["IMD"].astype(int)
age_imd_long["population"] = age_imd_long["population"].astype(int)

# Remove zero population cells
age_imd_long = age_imd_long[
    age_imd_long["population"] > 0
].reset_index(drop=True)

# Calculate sampling probabilities
probabilities = (
    age_imd_long["population"] /
    age_imd_long["population"].sum()
)

# Sample 50,000 women using the joint Age × IMD distribution
sampled_indices = rng.choice(
    len(age_imd_long),
    size=N,
    replace=True,
    p=probabilities
)

sampled_women = age_imd_long.iloc[sampled_indices].reset_index(drop=True)

# Create individual-level cohort
cohort = pd.DataFrame({
    "start_age": sampled_women["start_age"].values,
    "imd": sampled_women["IMD"].values,
    "n": 1.0
})

# Checks
print("Total cohort size:", len(cohort))
print("Total n:", cohort["n"].sum())

print("\nAge distribution:")
print(cohort["start_age"].value_counts().sort_index())

print("\nIMD distribution:")
print(cohort["imd"].value_counts().sort_index())

print("\nAge × IMD distribution:")
print(
    cohort.groupby(["imd", "start_age"])
    .size()
    .unstack(fill_value=0)
)

# 2. DISTRIBUTIONS
# ------------------

def dist_lognormal(mu_l: float, sigma_l: float):
    return lognorm(s=sigma_l, scale=np.exp(mu_l))

def dist_beta(a: float, b: float):
    return beta(a, b)

def dist_gamma(k: float, rate: float):
    return gamma(k, scale=1/rate)

def dist_dirichlet(alphas: List[float]):
    return dirichlet(alphas)

# --------------------------------------------------------
# 3. DECISION TREE
# --------------------------------------------------------
# 3.1 INPUTS
# ------------------

PARAMS: Dict[str, Dict[str, Union[float, Tuple, list]]] = {
    "COST_FHQS_TEXT":        dist_lognormal(3.091, 0.618),
    "COST_UPTAKE":           dist_gamma(32.83, 3.29),
    "COST_HIGH_RISK":        351.88,
    "COST_REFERRAL":         66.93,

    "P_FHQS_UPTAKE":         dist_beta(203, 1277),
    "P_RISK_SPLIT_FHQS":     dist_dirichlet([88.7, 10.3, 1.0]),

    "IMD_REFERRAL": {
        1:  dist_beta(5,   19224),
        2:  dist_beta(17,  56094),
        3:  dist_beta(24,  67475),
        4:  dist_beta(27,  67777),
        5:  dist_beta(40,  89535),
        6:  dist_beta(60, 119543),
        7:  dist_beta(58, 107788),
        8:  dist_beta(65, 111408),
        9:  dist_beta(71, 112480),
        10: dist_beta(60, 89261),
    },

    "P_RISK_SPLIT_REFERRAL":  dist_dirichlet([50.3, 40.4, 9.3]),
}

# 3.2 NODE CLASSES
# ------------------

@dataclass
class Node:

    def expected_cost(self) -> float:
        raise NotImplementedError

    def simulate(self, rng=None) -> Tuple[float, str]:
        raise NotImplementedError


@dataclass
class Terminal(Node):
    cost: float
    label: str = "TERMINAL"

    def expected_cost(self) -> float:
        return self.cost

    def simulate(self, rng=None) -> Tuple[float, str]:
        return self.cost, self.label


@dataclass
class Chance(Node):
    branches: List[Tuple[float, Node]]

    def expected_cost(self) -> float:
        return sum(
            probability * node.expected_cost()
            for probability, node in self.branches
        )

    def simulate(self, rng=None) -> Tuple[float, str]:

        probs = np.array(
            [probability for probability, _ in self.branches],
            dtype=float
        )

        probs = probs / probs.sum()

        if rng is None:
            rng = np.random.default_rng()

        choice = rng.choice(
            len(self.branches),
            p=probs
        )

        return self.branches[choice][1].simulate(rng)


@dataclass
class Decision(Node):
    options: Dict[str, Node]

    def expected_cost(self, option: str) -> float:
        return self.options[option].expected_cost()

    def simulate(self, option: str, rng=None) -> Tuple[float, str]:
        return self.options[option].simulate(rng)


@dataclass
class AddCost(Node):
    cost: float
    next_node: Node

    def expected_cost(self):
        return self.cost + self.next_node.expected_cost()

    def simulate(self, rng=None):
        c_next, lbl = self.next_node.simulate(rng)
        return self.cost + c_next, lbl

# 3.3 TREE BUILDING
# ------------------

def make_tree(
    imd_decile: int,
    dt_draws: dict
) -> Decision:

    # Terminal nodes
    unclassified_term = Terminal(
        0.0,
        label="Unclassified"
    )

    low_risk_term = Terminal(
        0.0,
        label="Low_risk"
    )

    mod_risk_term = Terminal(
        0.0,
        label="Moderate_risk"
    )

    high_risk_term = Terminal(
        dt_draws["COST_HIGH_RISK"],
        label="High_risk"
    )

    # FHQS risk split
    fhqs_risk_split = dt_draws["P_RISK_SPLIT_FHQS"]

    fhqs_risk_node = Chance([
        (
            fhqs_risk_split[0],
            low_risk_term
        ),
        (
            fhqs_risk_split[1],
            mod_risk_term
        ),
        (
            fhqs_risk_split[2],
            high_risk_term
        )
    ])
    # Routine care risk split following referral
    referral_risk_split = dt_draws["P_RISK_SPLIT_REFERRAL"]

    referral_risk_node = Chance([
        (
            referral_risk_split[0],
            low_risk_term
        ),
        (
            referral_risk_split[1],
            mod_risk_term
        ),
        (
            referral_risk_split[2],
            high_risk_term
        )
    ])
    # FHQS uptake probability
    p_uptake = dt_draws["P_FHQS_UPTAKE"]

    uptake_node = Chance([
        (
            p_uptake,
            AddCost(
                dt_draws["COST_UPTAKE"],
                fhqs_risk_node
            )
        ),
        (
            1 - p_uptake,
            unclassified_term
        )
    ])
    send_text_branch = AddCost(
        dt_draws["COST_FHQS_TEXT"],
        uptake_node
    )

    # Probability of self-referral
    p_referral = dt_draws["IMD_REFERRAL"][imd_decile]

    referral_node = Chance([
        (
            p_referral,
            AddCost(
                dt_draws["COST_REFERRAL"],
                referral_risk_node
            )
        ),
        (
            1 - p_referral,
            unclassified_term
        )
    ])
    # Seperate intervention and control arms
    return Decision({
        "SEND_TEXT": send_text_branch,
        "NO_TEXT": referral_node
    })

# 3.4. SIMULATE COHORT FUNCTION
# -----------------------------

def simulate_cohort_separate_risks(cohort: pd.DataFrame, dt_draws: dict, rng):
    # Simulate each woman once through the tree for both arms.
    # Returns total and by-IMD statistics plus individual-level data.

    imd_levels = range(1, 11)

    total_control = {
        "cost": 0.0,
        "Unclassified": 0,
        "Low_risk": 0,
        "Moderate_risk": 0,
        "High_risk": 0,
    }

    total_interv = {
        "cost": 0.0,
        "Unclassified": 0,
        "Low_risk": 0,
        "Moderate_risk": 0,
        "High_risk": 0,
    }

    results_control_by_imd = {
        imd: {
            "cost": 0.0,
            "Unclassified": 0,
            "Low_risk": 0,
            "Moderate_risk": 0,
            "High_risk": 0
        }
        for imd in imd_levels
    }

    results_interv_by_imd = {
        imd: {
            "cost": 0.0,
            "Unclassified": 0,
            "Low_risk": 0,
            "Moderate_risk": 0,
            "High_risk": 0
        }
        for imd in imd_levels
    }

    # Individual-level data
    indiv_data_control = []
    indiv_data_interv = []

    # ---------------------------------------------------------
    # SIMULATE EACH WOMAN ONCE
    # ---------------------------------------------------------

    for _, person in cohort.iterrows():

        imd = int(person["imd"])
        start_age = int(person["start_age"])

        # Build tree using the PSA parameter draws
        tree = make_tree(
            imd_decile=imd,
            dt_draws=dt_draws
        )

        # CONTROL ARM

        cost_c, label_c = tree.simulate(
            "NO_TEXT",
            rng=rng
        )

        total_control["cost"] += cost_c
        total_control[label_c] += 1

        results_control_by_imd[imd]["cost"] += cost_c
        results_control_by_imd[imd][label_c] += 1

        indiv_data_control.append({
            "start_age": start_age,
            "imd": imd,
            "risk": label_c.lower()
        })

        # INTERVENTION ARM

        cost_i, label_i = tree.simulate(
            "SEND_TEXT",
            rng=rng
        )

        total_interv["cost"] += cost_i
        total_interv[label_i] += 1

        results_interv_by_imd[imd]["cost"] += cost_i
        results_interv_by_imd[imd][label_i] += 1

        indiv_data_interv.append({
            "start_age": start_age,
            "imd": imd,
            "risk": label_i.lower()
        })

    # CONVERT TO DATAFRAMES FOR MARKOV MODELS

    indiv_data_control_df = pd.DataFrame(
        indiv_data_control
    )

    indiv_data_interv_df = pd.DataFrame(
        indiv_data_interv
    )

    return (
        total_control,
        results_control_by_imd,
        total_interv,
        results_interv_by_imd,
        indiv_data_control_df,
        indiv_data_interv_df
    )


# -------------------------------------------------------------------
# 4. MARKOV MODEL 1 FOR GENERAL POPULATION
# -------------------------------------------------------------------
# 4.1 MODEL SETUP
# -----------------------
MODEL1_STATES = [
    "BC_free_unclassified",
    "Symptomatic_detection",
    "Low_risk",
    "Moderate_risk",
    "High_risk"
]

MODEL1_ALL_STATES = MODEL1_STATES + ["Exited"]

AGE_RANGE = range(38, 50) # Defining age range
CONF_LEVEL = 0.95

# 4.2 COST PARAMETERS
# ----------------------

# Costs incurred when transitioning into a risk category
transition_costs = {
    "to_Low_risk": 66.93,
    "to_Moderate_risk": 66.93,
    "to_High_risk": 418.81
}

# Cost of symptomatic detection
# Gamma distribution parameterised by shape (alpha) and rate
symptomatic_detection_cost = dist_gamma(894.8, 0.780)

# 4.3 Probabilities
# -----------------------
# prevalence_df = pd.read_excel("/content/drive/My Drive/Colab Notebooks/Markov prevelance table two.xlsx")
# prevalence_df.columns = prevalence_df.columns.str.strip()  # Strip spaces from column names

symptomatic_detection_probs = {
    38: dist_beta(6.7, 9993.3),
    39: dist_beta(6.7, 9993.3),
    40: dist_beta(12.7, 9987.3),
    41: dist_beta(12.7, 9987.3),
    42: dist_beta(12.7, 9987.3),
    43: dist_beta(12.7, 9987.3),
    44: dist_beta(12.7, 9987.3),
    45: dist_beta(21.4, 9978.6),
    46: dist_beta(21.4, 9978.6),
    47: dist_beta(21.4, 9978.6),
    48: dist_beta(21.4, 9978.6),
    49: dist_beta(21.4, 9978.6)
}
IMD_referral_probs = {
    1: dist_beta(5, 19224),
    2: dist_beta(17, 56094),
    3: dist_beta(24, 67475),
    4: dist_beta(27, 67777),
    5: dist_beta(40, 89535),
    6: dist_beta(60, 119543),
    7: dist_beta(58, 107788),
    8: dist_beta(65, 111408),
    9: dist_beta(71, 112480),
    10: dist_beta(60, 89261)
}

# Probability of risk category conditional on genetics referral
P_RISK_SPLIT_REFERRAL = dist_dirichlet([50.3, 40.4, 9.3])

# 4.4 TRANSITION PROBABILITIES
# ---------------------------------------------------------------

def unclassified_transitions(
    age,
    imd,
    referral_draw,
    risk_split_draw,
    symptomatic_draw
):

    # Annual symptomatic detection probability
    p_symptomatic = symptomatic_draw

    # Annual probability of self-referral
    p_referral = referral_draw

    # Risk category following referral
    p_low_risk = p_referral * risk_split_draw[0]
    p_moderate_risk = p_referral * risk_split_draw[1]
    p_high_risk = p_referral * risk_split_draw[2]

    # Remain unclassified
    p_remain = (
        1
        - p_symptomatic
        - p_low_risk
        - p_moderate_risk
        - p_high_risk
    )

    return {
        "BC_free_unclassified": p_remain,
        "Symptomatic_detection": p_symptomatic,
        "Low_risk": p_low_risk,
        "Moderate_risk": p_moderate_risk,
        "High_risk": p_high_risk
    }


def low_risk_transitions(age, imd, symptomatic_draw):

    p_symptomatic = symptomatic_draw

    return {
        "BC_free_unclassified": 0,
        "Symptomatic_detection": p_symptomatic,
        "Low_risk": 1 - p_symptomatic,
        "Moderate_risk": 0,
        "High_risk": 0
    }


def build_transition_matrix(
    age,
    imd,
    referral_draw,
    risk_split_draw,
    symptomatic_draw
):

    # Get transitions from unclassified state
    unclassified = unclassified_transitions(
        age=age,
        imd=imd,
        referral_draw=referral_draw,
        risk_split_draw=risk_split_draw,
        symptomatic_draw=symptomatic_draw
    )

    # Get transitions from low-risk state
    low_risk = low_risk_transitions(
        age=age,
        imd=imd,
        symptomatic_draw=symptomatic_draw
    )

    # Transition matrix
    M = pd.DataFrame(
        0.0,
        index=MODEL1_ALL_STATES,
        columns=MODEL1_ALL_STATES
    )

    # BC-free unclassified
    M.loc[
        "BC_free_unclassified",
        "BC_free_unclassified"
    ] = unclassified["BC_free_unclassified"]

    M.loc[
        "BC_free_unclassified",
        "Symptomatic_detection"
    ] = unclassified["Symptomatic_detection"]

    M.loc[
        "BC_free_unclassified",
        "Low_risk"
    ] = unclassified["Low_risk"]

    M.loc[
        "BC_free_unclassified",
        "Moderate_risk"
    ] = unclassified["Moderate_risk"]

    M.loc[
        "BC_free_unclassified",
        "High_risk"
    ] = unclassified["High_risk"]

    # Low-risk
    M.loc[
        "Low_risk",
        "Low_risk"
    ] = low_risk["Low_risk"]

    M.loc[
        "Low_risk",
        "Symptomatic_detection"
    ] = low_risk["Symptomatic_detection"]

    # Exit states
    M.loc["Moderate_risk", "Exited"] = 1.0
    M.loc["High_risk", "Exited"] = 1.0
    M.loc["Symptomatic_detection", "Exited"] = 1.0

    # Exited remains exited
    M.loc["Exited", "Exited"] = 1.0

    return M


# 4.5 MODEL 1 SIMULATION FUNCTION
# ---------------------------------------------------------------

def run_markov_model(
    cohort,
    arm,
    referral_draws,
    risk_split_draw,
    symptomatic_detection_draws,
    symptomatic_cost_draw
):

    results = []
    risk_identifications = []

    total_cost = 0.0
    total_symptomatic = 0.0

    # Group by starting age, IMD and initial risk
    if "n" in cohort.columns:
    # Deterministic model: fractional population counts
      grouped = (
        cohort
        .groupby(["start_age", "imd", "risk"])["n"]
        .sum()
        .reset_index()
      )

    else:
      # PSA / individual-level cohort: each row = 1 woman
      grouped = (
        cohort
        .groupby(["start_age", "imd", "risk"])
        .size()
        .reset_index(name="n")
      )

    for _, row in grouped.iterrows():

        start_age = int(row["start_age"])
        imd = int(row["imd"])
        initial_risk = row["risk"]
        n = row["n"]

        # INITIAL STATE

        population = pd.Series(
            0.0,
            index=MODEL1_ALL_STATES
        )

        if initial_risk == "unclassified":

            population["BC_free_unclassified"] = n

        elif initial_risk == "low_risk":

            population["Low_risk"] = n

        # FOLLOW FROM STARTING AGE TO AGE 49

        for age in range(start_age, 50):

            M = build_transition_matrix(
                age=age,
                imd=imd,
                referral_draw=referral_draws[imd],
                risk_split_draw=risk_split_draw,
                symptomatic_draw=symptomatic_detection_draws[age]
            )

            # RISK IDENTIFICATION

            n_low = (
                population["BC_free_unclassified"]
                * M.loc[
                    "BC_free_unclassified",
                    "Low_risk"
                ]
            )

            n_moderate = (
                population["BC_free_unclassified"]
                * M.loc[
                    "BC_free_unclassified",
                    "Moderate_risk"
                ]
            )

            n_high = (
                population["BC_free_unclassified"]
                * M.loc[
                    "BC_free_unclassified",
                    "High_risk"
                ]
            )

            # Store moderate/high-risk women who will enter Model 2
            if n_moderate > 0:

                risk_identifications.append({
                    "arm": arm,
                    "age": age,
                    "imd": imd,
                    "risk": "moderate_risk",
                    "n": n_moderate
                })

            if n_high > 0:

                risk_identifications.append({
                    "arm": arm,
                    "age": age,
                    "imd": imd,
                    "risk": "high_risk",
                    "n": n_high
                })

            # SYMPTOMATIC DETECTION

            symptomatic_unclassified = (
                population["BC_free_unclassified"]
                * M.loc[
                    "BC_free_unclassified",
                    "Symptomatic_detection"
                ]
            )

            symptomatic_low_risk = (
                population["Low_risk"]
                * M.loc[
                    "Low_risk",
                    "Symptomatic_detection"
                ]
            )

            symptomatic_this_year = (
                symptomatic_unclassified
                + symptomatic_low_risk
            )

            # COSTS COUNTS

            cost_low = (
                n_low
                * transition_costs["to_Low_risk"]
            )

            cost_moderate = (
                n_moderate
                * transition_costs["to_Moderate_risk"]
            )

            cost_high = (
                n_high
                * transition_costs["to_High_risk"]
            )

            cost_symptomatic = (
                symptomatic_this_year
                * symptomatic_cost_draw
            )

            cost_this_year = (
                cost_low
                + cost_moderate
                + cost_high
                + cost_symptomatic
            )

            total_cost += cost_this_year
            total_symptomatic += symptomatic_this_year

            # STORE RESULTS

            results.append({
                "arm": arm,
                "age": age,
                "imd": imd,
                "start_age": start_age,
                "initial_risk": initial_risk,
                "n": n,
                "low_identified": n_low,
                "moderate_identified": n_moderate,
                "high_identified": n_high,
                "symptomatic_detected": symptomatic_this_year,
                "cost": cost_this_year
            })

            # MOVE POPULATION FORWARD

            population = population @ M

    results_df = pd.DataFrame(results)

    risk_identifications_df = pd.DataFrame(
        risk_identifications,
        columns=[
            "arm",
            "age",
            "imd",
            "risk",
            "n"
        ]
    )

    return (
        results_df,
        risk_identifications_df,
        total_cost,
        total_symptomatic
    )

# ---------------------------------------------
# 5. MARKOV MODEL 2: MOD/HIGH RISK
# ----------------------------------------------
import numpy as np
import pandas as pd
from scipy.stats import beta, gamma

# 5.1 MODEL SETUP
# ----------------
MODEL2_STATES = [
    "Healthy_TN",
    "Healthy_FP",
    "Cancer_Undetected",
    "Cancer_Detected"
]

MODEL2_ALL_STATES = MODEL2_STATES + ["Exited"]
RISK_GROUPS = ["high_risk", "moderate_risk"]
AGE_RANGE = range(38, 50)
CONF_LEVEL = 0.95

# Sensitivity and Specificity
SENS_MEAN, SENS_L, SENS_U = 0.780, 0.730, 0.830
SPEC_MEAN, SPEC_L, SPEC_U = 0.90, 0.897, 0.903
UP_MEAN, UP_L, UP_U = 0.824, 0.754, 0.894

# 5.2 COST PARAMETERS
# -----------------
cost_params = {
    "Healthy_TN": 196.40,
    "Healthy_FP": dist_gamma(207.78, 0.376),
    "Cancer_Undetected": 196.40,
    "Cancer_Detected": dist_gamma(536, 0.604)
}

# Beta fitting
from scipy.stats import beta, gamma, norm

def beta_from_ci(mean, low, high, conf=CONF_LEVEL):
    z = norm.ppf(1 - (1 - conf) / 2)
    sd = (high - low) / (2 * z)
    var = sd**2

    tmp = mean * (1 - mean) / var - 1

    alpha = mean * tmp
    beta_param = (1 - mean) * tmp

    return beta(alpha, beta_param)

sens_dist = beta_from_ci(SENS_MEAN, SENS_L, SENS_U)
spec_dist = beta_from_ci(SPEC_MEAN, SPEC_L, SPEC_U)
uptake_dist = beta_from_ci(UP_MEAN, UP_L, UP_U)

# 5.3 LOAD PREVALENCE DATA
# ----------------------
prevalence_df = pd.read_excel("/content/drive/My Drive/Colab Notebooks/Markov prevelance table.xlsx")
prevalence_df.columns = prevalence_df.columns.str.strip()
def load_prevalence_table(path):
    df = pd.read_excel(path)
    df.columns = df.columns.str.strip()
    return df

# 5.4 HELPER FUNCTIONS
# -----------------------
def band_for_age(age):
    return "38-39" if age < 40 else ("40-44" if age < 45 else "45-49")

def make_matrix(prev, sens, spec, uptake):
    tn = (1 - prev) * spec
    fp = (1 - prev) * (1 - spec)
    fn = prev * (1 - sens)
    tp = prev * sens
    p = {"TN": tn, "FP": fp, "FN": fn, "TP": tp, "sens": sens}

    M = pd.DataFrame(0., index=MODEL2_ALL_STATES, columns=MODEL2_ALL_STATES)
    M.loc["Healthy_TN", "Healthy_TN"] = (1 - uptake) + uptake * p["TN"]
    M.loc["Healthy_TN", "Healthy_FP"] = uptake * p["FP"]
    M.loc["Healthy_TN", "Cancer_Undetected"] = uptake * p["FN"]
    M.loc["Healthy_TN", "Cancer_Detected"] = uptake * p["TP"]

    M.loc["Healthy_FP", "Healthy_FP"] = (1 - uptake) + uptake * p["FP"]
    M.loc["Healthy_FP", "Healthy_TN"] = uptake * p["TN"]
    M.loc["Healthy_FP", "Cancer_Undetected"] = uptake * p["FN"]
    M.loc["Healthy_FP", "Cancer_Detected"] = uptake * p["TP"]

    M.loc["Cancer_Undetected", "Cancer_Undetected"] = (1 - uptake) + uptake * (1 - p["sens"])
    M.loc["Cancer_Undetected", "Cancer_Detected"] = uptake * p["sens"]

    M.loc["Cancer_Detected", "Exited"] = 1.0
    M.loc["Exited", "Exited"] = 1.0
    return M


# 5.5 MATRIX BUILDER
# ------------------
def build_matrices(prevalence_df, sens_draw, spec_draw, uptake_draw):

    matrices = {}

    for _, row in prevalence_df.iterrows():

        risk = row["risk"]
        age_band = row["age_band"]
        prevalence = float(row["prevelance"])

        matrices[(risk, age_band)] = make_matrix(
            prevalence,
            sens_draw,
            spec_draw,
            uptake_draw
        )

    return matrices

# 5.6 CANCER DETECTION FUNCTION
# --------------------------
def cancers_detected_before_50_grouped(matrices, grouped_df):

    results = []

    for _, row in grouped_df.iterrows():

        risk = row["risk"]
        start_age = int(row["start_age"])
        imd = int(row["imd"])
        n = float(row["n"])

        dist = pd.Series(0.0, index=MODEL2_ALL_STATES)
        dist["Healthy_TN"] = n

        for age in range(start_age, 50):
            age_band = band_for_age(age)
            dist = dist @ matrices[(risk, age_band)]

        results.append({
            "imd": imd,
            "risk": risk,
            "start_age": start_age,
            "n": n,
            "detected": dist["Exited"]
        })

    return pd.DataFrame(results)


# 5.7 COST CALCULATOR
# -----------------
def calculate_costs(matrices, grouped_df, cost_draws, discount_rate=0.035):

    results = []

    for _, row in grouped_df.iterrows():

        risk = row["risk"]
        start_age = int(row["start_age"])
        imd = int(row["imd"])
        n = float(row["n"])

        dist = pd.Series(0.0, index=MODEL2_ALL_STATES)
        dist["Healthy_TN"] = n

        cohort_cost = 0.0

        for i, age in enumerate(range(start_age, 50)):

            age_band = band_for_age(age)
            M = matrices[(risk, age_band)]

            year_cost = (
                dist["Healthy_TN"] * cost_draws["Healthy_TN"] +
                dist["Healthy_FP"] * cost_draws["Healthy_FP"] +
                dist["Cancer_Undetected"] * cost_draws["Cancer_Undetected"] +
                dist["Cancer_Detected"] * cost_draws["Cancer_Detected"]
            )

            cohort_cost += year_cost / ((1 + discount_rate) ** i)

            dist = dist @ M

        results.append({
            "imd": imd,
            "risk": risk,
            "start_age": start_age,
            "n": n,
            "cost": cohort_cost
        })

    return pd.DataFrame(results)

# 5.8 MODEL 2 SIMULATION FUNCTION
# --------------------------------------------

def run_model2(patients_df, cost_draws, matrices):

    required_cols = ["risk", "start_age", "imd", "n"]

    # Check that the required columns exist
    missing_cols = [
        col for col in required_cols
        if col not in patients_df.columns
    ]

    if missing_cols:
        raise ValueError(
            f"run_model2 is missing columns: {missing_cols}"
        )

    # Select only the columns we need
    patients_df = patients_df[
        ["risk", "start_age", "imd", "n"]
    ].copy()

    # Make sure n is numeric
    patients_df["n"] = pd.to_numeric(
        patients_df["n"],
        errors="raise"
    )

    # Group the cohort
    grouped = (
        patients_df
        .groupby(
            ["risk", "start_age", "imd"],
            as_index=False
        )
        .sum(numeric_only=True)
    )

    # Keep only the variables needed downstream
    grouped = grouped[
        ["risk", "start_age", "imd", "n"]
    ]

    # Check n explicitly
    if "n" not in grouped.columns:
        raise ValueError(
            "Grouped Model 2 cohort does not contain 'n'"
        )

    # EFFECT

    effect = cancers_detected_before_50_grouped(
        matrices,
        grouped
    )

    cost = calculate_costs(
        matrices,
        grouped,
        cost_draws
    )

    return cost, effect


# ----------------------------------------
# 6. PROBABILISTIC SENSITIVITY ANALYSIS (PSA)
# ----------------------------------------

if __name__ == "__main__":
    N_PSA = 1 # Number of simulation
    imd_levels = range(1, 11)

    # PSA result containers for decision tree
    psa_totals = {
    "control_cost": [],
    "interv_cost": [],

    "control_no_bc": [],
    "interv_no_bc": [],

    "control_screen_detected": [],
    "interv_screen_detected": [],

    "control_symptomatic_detections": [],
    "interv_symptomatic_detections": [],

    "control_moderate_risk": [],
    "interv_moderate_risk": [],

    "control_high_risk": [],
    "interv_high_risk": [],

    "control_moderate_risk_identified": [],
    "interv_moderate_risk_identified": [],

    "control_high_risk_identified": [],
    "interv_high_risk_identified": [],
    }
    psa_by_imd = {
        key: {imd: [] for imd in imd_levels}
        for key in psa_totals.keys()
    }

    # PSA result containers for markov
    markov_results = {
        "control_cost": [],
        "control_effects": [],
        "control_cost_per_case": [],
        "interv_cost": [],
        "interv_effects": [],
        "interv_cost_per_case": [],
    }

    # PSA result containers for markov by IMD

    markov_results_by_imd = {
        imd: {
            "control_cost": [],
            "control_effects": [],
            "control_cost_per_case": [],
            "interv_cost": [],
            "interv_effects": [],
            "interv_cost_per_case": [],
        }
        for imd in imd_levels
    }


    excel_path = ("/content/drive/My Drive/Colab Notebooks/Markov prevelance table.xlsx")

    for i in range(N_PSA):

      print(f"PSA iteration {i+1}/{N_PSA}")

      ng = np.random.default_rng()

      # DRAW UNCERTAINTY FOR DECISION TREE

      dt_draws = {
        "COST_FHQS_TEXT": PARAMS["COST_FHQS_TEXT"].rvs(random_state=ng),
        "COST_UPTAKE": PARAMS["COST_UPTAKE"].rvs(random_state=ng),
        "COST_HIGH_RISK": PARAMS["COST_HIGH_RISK"],
        "COST_REFERRAL": PARAMS["COST_REFERRAL"],

        "P_FHQS_UPTAKE": PARAMS["P_FHQS_UPTAKE"].rvs(random_state=ng),

        "P_RISK_SPLIT_FHQS": PARAMS["P_RISK_SPLIT_FHQS"].rvs(
           random_state=ng
        )[0],

        "IMD_REFERRAL": {
           imd: PARAMS["IMD_REFERRAL"][imd].rvs(random_state=ng)
           for imd in imd_levels
        },

        "P_RISK_SPLIT_REFERRAL": PARAMS["P_RISK_SPLIT_REFERRAL"].rvs(
           random_state=ng
        )[0],
      }

      # DRAW UNCERTAINTY FOR MARKOV MODEL 1

      # IMD-specific referral probabilities
      referral_draws = {
        imd: IMD_referral_probs[imd].rvs(random_state=ng)
        for imd in imd_levels
      }

      # Risk split following referral
      risk_split_draw = P_RISK_SPLIT_REFERRAL.rvs(
        random_state=ng
      )[0]

      # Age-specific symptomatic detection probabilities
      symptomatic_detection_draws = {
        age: symptomatic_detection_probs[age].rvs(
            random_state=ng
        )
        for age in AGE_RANGE
      }

      # Symptomatic detection cost
      symptomatic_cost_draw = symptomatic_detection_cost.rvs(
        random_state=ng
      )

      # DRAW UNCERTAINTY FOR MARKOV MODEL 2

      sens_draw = sens_dist.rvs(random_state=ng)
      spec_draw = spec_dist.rvs(random_state=ng)
      uptake_draw = uptake_dist.rvs(random_state=ng)

      model2_cost_draws = {
        "Healthy_TN": cost_params["Healthy_TN"],
        "Healthy_FP": cost_params["Healthy_FP"].rvs(
            random_state=ng
        ),
        "Cancer_Undetected": cost_params["Cancer_Undetected"],
        "Cancer_Detected": cost_params["Cancer_Detected"].rvs(
            random_state=ng
        )
      }
      # Build Model 2 transition matrices using
      # the SAME parameter draws for both arms
      model2_matrices = build_matrices(
        prevalence_df,
        sens_draw,
        spec_draw,
        uptake_draw
      )

      # SIMULATE DECISION TREE

      (
         total_control,
         results_control_by_imd,
         total_interv,
         results_interv_by_imd,
         indiv_data_control_df,
         indiv_data_interv_df
      ) = simulate_cohort_separate_risks(
          cohort=cohort,
          dt_draws=dt_draws,
          rng=ng
      )

      # MARKOV MODEL 1 COHORT

      # Unclassified + low-risk women from decision tree

      control_model1_cohort = (
        indiv_data_control_df[
            indiv_data_control_df["risk"].isin(
                ["unclassified", "low_risk"]
            )
        ].copy()
      )

      interv_model1_cohort = (
        indiv_data_interv_df[
            indiv_data_interv_df["risk"].isin(
                ["unclassified", "low_risk"]
            )
        ].copy()
      )

      # SIMULATE MARKOV MODEL 1

      (
        control_model1_results,
        control_model1_identified,
        control_model1_cost,
        control_model1_symptomatic
      ) = run_markov_model(
        cohort=control_model1_cohort,
        arm="control",
        referral_draws=referral_draws,
        risk_split_draw=risk_split_draw,
        symptomatic_detection_draws=symptomatic_detection_draws,
        symptomatic_cost_draw=symptomatic_cost_draw
      )

      (
        interv_model1_results,
        interv_model1_identified,
        interv_model1_cost,
        interv_model1_symptomatic
      ) = run_markov_model(
        cohort=interv_model1_cohort,
        arm="intervention",
        referral_draws=referral_draws,
        risk_split_draw=risk_split_draw,
        symptomatic_detection_draws=symptomatic_detection_draws,
        symptomatic_cost_draw=symptomatic_cost_draw
      )

      # MARKOV MODEL 2 COHORT FROM DECISION TREE

      control_original_model2 = (
       indiv_data_control_df[
        indiv_data_control_df["risk"].isin(
            ["moderate_risk", "high_risk"]
        )
       ]
      .groupby(["risk", "start_age", "imd"])
      .size()
      .reset_index(name="n")
      )

      interv_original_model2 = (
       indiv_data_interv_df[
        indiv_data_interv_df["risk"].isin(
            ["moderate_risk", "high_risk"]
        )
       ]
      .groupby(["risk", "start_age", "imd"])
      .size()
      .reset_index(name="n")
      )

      # MARKOV MODEL 2 COHORT FROM MODEL 1

      control_new_model2 = (
         control_model1_identified[
            control_model1_identified["risk"].isin(
              ["moderate_risk", "high_risk"]
            )
         ]
         .rename(columns={"age": "start_age"})
         .groupby(["risk", "start_age", "imd"])["n"]
         .sum()
         .reset_index()
      )

      interv_new_model2 = (
        interv_model1_identified[
          interv_model1_identified["risk"].isin(
            ["moderate_risk", "high_risk"]
          )
        ]
        .rename(columns={"age": "start_age"})
        .groupby(["risk", "start_age", "imd"])["n"]
        .sum()
        .reset_index()
      )

      # COMBINE MARKOV MODEL 2 ENTRANTS

      control_markov2_cohort = pd.concat(
        [
            control_original_model2,
            control_new_model2
        ],
        ignore_index=True
      )

      interv_markov2_cohort = pd.concat(
        [
            interv_original_model2,
            interv_new_model2
        ],
        ignore_index=True
      )

      # STORE MODERATE- AND HIGH-RISK IDENTIFICATION: Decision tree + additional Model 1 identification

      # Control arm
      control_moderate_risk_identified = (
         control_original_model2.loc[
            control_original_model2["risk"] == "moderate_risk", "n"
         ].sum()
         + control_new_model2.loc[
             control_new_model2["risk"] == "moderate_risk", "n"
         ].sum()
      )

      control_high_risk_identified = (
         control_original_model2.loc[
            control_original_model2["risk"] == "high_risk", "n"
         ].sum()
         + control_new_model2.loc[
            control_new_model2["risk"] == "high_risk", "n"
         ].sum()
      )

      # Intervention arm
      interv_moderate_risk_identified = (
          interv_original_model2.loc[
             interv_original_model2["risk"] == "moderate_risk", "n"
          ].sum()
          + interv_new_model2.loc[
             interv_new_model2["risk"] == "moderate_risk", "n"
          ].sum()
      )

      interv_high_risk_identified = (
          interv_original_model2.loc[
             interv_original_model2["risk"] == "high_risk", "n"
          ].sum()
          + interv_new_model2.loc[
             interv_new_model2["risk"] == "high_risk", "n"
          ].sum()
      )

      # Append results for this PSA iteration
      psa_totals["control_moderate_risk_identified"].append(
         control_moderate_risk_identified
      )
      psa_totals["interv_moderate_risk_identified"].append(
         interv_moderate_risk_identified
      )

      psa_totals["control_high_risk_identified"].append(
         control_high_risk_identified
      )
      psa_totals["interv_high_risk_identified"].append(
         interv_high_risk_identified
      )

      # SIMULATE MARKOV MODEL 2

      control_model2_cost, control_model2_effect = run_model2(
        control_markov2_cohort,
        model2_cost_draws,
        model2_matrices
      )

      interv_model2_cost, interv_model2_effect = run_model2(
        interv_markov2_cohort,
        model2_cost_draws,
        model2_matrices
      )

      # TOTAL COST AND EFFECT

      # Model 2: aggregate costs

      control_model2_total_cost = control_model2_cost["cost"].sum()
      interv_model2_total_cost = interv_model2_cost["cost"].sum()

      # Model 2: aggregate effects

      control_model2_total_effect = control_model2_effect["detected"].sum()
      interv_model2_total_effect = interv_model2_effect["detected"].sum()

      # Total costs

      control_total_cost = (
        total_control["cost"]
        + control_model1_cost
        + control_model2_total_cost
      )

      interv_total_cost = (
        total_interv["cost"]
        + interv_model1_cost
        + interv_model2_total_cost
      )

      # Total effects

      # Symptomatic detections come from Model 1
      control_total_symptomatic = control_model1_symptomatic
      interv_total_symptomatic = interv_model1_symptomatic

      # Screen-detected cases come from Model 2
      control_total_screen_detected = control_model2_total_effect
      interv_total_screen_detected = interv_model2_total_effect

      # STORE PSA RESULTS

      psa_totals["control_cost"].append(
        control_total_cost
      )

      psa_totals["interv_cost"].append(
        interv_total_cost
      )

      psa_totals["control_symptomatic_detections"].append(
        control_total_symptomatic
      )

      psa_totals["interv_symptomatic_detections"].append(
        interv_total_symptomatic
      )

      psa_totals["control_screen_detected"].append(
        control_total_screen_detected
      )

      psa_totals["interv_screen_detected"].append(
        interv_total_screen_detected
      )

      # PSA SUMMARY OUTPUTS

    summary = pd.DataFrame({
      "Outcome": [
           "Total cost - Control",
           "Total cost - Intervention",
           "Symptomatic detections - Control",
           "Symptomatic detections - Intervention",
           "Screen-detected - Control",
           "Screen-detected - Intervention",
           "Moderate-risk identified - Control",
           "Moderate-risk identified - Intervention",
           "High-risk identified - Control",
           "High-risk identified - Intervention"
      ],
      "Mean": [
           np.mean(psa_totals["control_cost"]),
           np.mean(psa_totals["interv_cost"]),
           np.mean(psa_totals["control_symptomatic_detections"]),
           np.mean(psa_totals["interv_symptomatic_detections"]),
           np.mean(psa_totals["control_screen_detected"]),
           np.mean(psa_totals["interv_screen_detected"]),
           np.mean(psa_totals["control_moderate_risk_identified"]),
           np.mean(psa_totals["interv_moderate_risk_identified"]),
           np.mean(psa_totals["control_high_risk_identified"]),
           np.mean(psa_totals["interv_high_risk_identified"]),
      ],
      "Lower_95_CI": [
           np.percentile(psa_totals["control_cost"], 2.5),
           np.percentile(psa_totals["interv_cost"], 2.5),
           np.percentile(psa_totals["control_symptomatic_detections"], 2.5),
           np.percentile(psa_totals["interv_symptomatic_detections"], 2.5),
           np.percentile(psa_totals["control_screen_detected"], 2.5),
           np.percentile(psa_totals["interv_screen_detected"], 2.5),
           np.percentile(psa_totals["control_moderate_risk_identified"], 2.5),
           np.percentile(psa_totals["interv_moderate_risk_identified"], 2.5),
           np.percentile(psa_totals["control_high_risk_identified"], 2.5),
           np.percentile(psa_totals["interv_high_risk_identified"], 2.5),
      ],
      "Upper_95_CI": [
           np.percentile(psa_totals["control_cost"], 97.5),
           np.percentile(psa_totals["interv_cost"], 97.5),
           np.percentile(psa_totals["control_symptomatic_detections"], 97.5),
           np.percentile(psa_totals["interv_symptomatic_detections"], 97.5),
           np.percentile(psa_totals["control_screen_detected"], 97.5),
           np.percentile(psa_totals["interv_screen_detected"], 97.5),
           np.percentile(psa_totals["control_moderate_risk_identified"], 97.5),
           np.percentile(psa_totals["interv_moderate_risk_identified"], 97.5),
           np.percentile(psa_totals["control_high_risk_identified"], 97.5),
           np.percentile(psa_totals["interv_high_risk_identified"], 97.5),
      ]
    })

      # Formatting
    summary["Mean"] = summary["Mean"].round(2)
    summary["Lower_95_CI"] = summary["Lower_95_CI"].round(2)
    summary["Upper_95_CI"] = summary["Upper_95_CI"].round(2)

    print(summary.to_string(index=False))

    # CALCULATING NET EFFECTS

    # Convert PSA lists to numeric NumPy arrays
    control_screen_detected = np.asarray(
      psa_totals["control_screen_detected"],
      dtype=float
    )

    control_symptomatic = np.asarray(
      psa_totals["control_symptomatic_detections"],
      dtype=float
    )

    interv_screen_detected = np.asarray(
      psa_totals["interv_screen_detected"],
      dtype=float
    )

    interv_symptomatic = np.asarray(
      psa_totals["interv_symptomatic_detections"],
      dtype=float
    )

    # NET EFFECT PER ITERATION

    Net_effect_control_sample = (
      control_screen_detected
      - control_symptomatic
    )

    Net_effect_interv_sample = (
      interv_screen_detected
      - interv_symptomatic
    )

    # MEAN NET EFFECT

    Net_effect_control = np.mean(
       Net_effect_control_sample
    )

    Net_effect_interv = np.mean(
       Net_effect_interv_sample
    )

    # 95% uncertainty intervals

    Net_effect_control_ui = np.percentile(
      Net_effect_control_sample,
      [2.5, 97.5]
    )

    Net_effect_interv_ui = np.percentile(
      Net_effect_interv_sample,
      [2.5, 97.5]
    )

    print(
      f"\nMean Net Effect Control (Screen-detected - Symptomatic): "
      f"{Net_effect_control:.4f} "
      f"(95% UI: "
      f"{Net_effect_control_ui[0]:.4f}–"
      f"{Net_effect_control_ui[1]:.4f})"
    )

    print(
      f"\nMean Net Effect Intervention (Screen-detected - Symptomatic): "
      f"{Net_effect_interv:.4f} "
      f"(95% UI: "
      f"{Net_effect_interv_ui[0]:.4f}–"
      f"{Net_effect_interv_ui[1]:.4f})"
    )

    # CALCULATING ICER

    # Incremental net effect for EACH PSA iteration
    Incremental_net_effect = Net_effect_interv - Net_effect_control
    Incremental_cost = np.mean(psa_totals["interv_cost"]) - np.mean(psa_totals["control_cost"])
    # Print incremental results
    print(
      f"\nIncremental Net Effect (Intervention - Control): "
      f"{Incremental_net_effect:.4f} "
    )
    print(
      f"\nIncremental costs: £{Incremental_cost:,.2f} "
    )
    # Calculating ICER
    ICER = (
       Incremental_cost / Incremental_net_effect
       if Incremental_net_effect != 0
       else float('inf')
    )
    print(
        f"ICER: {ICER}"
    )

    # CALCULATING TOTAL COSTS AND NET EFFECTS PER PSA SAMPLE

    # Total costs for each PSA sample
    Interv_costs_sample = np.array(psa_totals["interv_cost"])
    Control_costs_sample = np.array(psa_totals["control_cost"])

    # Incremental cost and effect for each PSA sample
    Incremental_costs_sample = Interv_costs_sample - Control_costs_sample
    Incremental_net_effects_sample = Net_effect_interv_sample - Net_effect_control_sample

    # PSA COST-EFFECTIVENESS PLANE QUADRANTS

    n_iterations = len(Incremental_costs_sample)

    # NE: More effective, more costly
    ne_quadrant = (
      (Incremental_net_effects_sample > 0) &
      (Incremental_costs_sample > 0)
    )

    # SE: More effective, less costly
    se_quadrant = (
      (Incremental_net_effects_sample > 0) &
      (Incremental_costs_sample < 0)
    )

    # NW: Less effective, more costly
    nw_quadrant = (
      (Incremental_net_effects_sample < 0) &
      (Incremental_costs_sample > 0)
    )

    # SW: Less effective, less costly
    sw_quadrant = (
      (Incremental_net_effects_sample < 0) &
      (Incremental_costs_sample < 0)
    )

    # Count PSA iterations in each quadrant
    n_ne = np.sum(ne_quadrant)
    n_se = np.sum(se_quadrant)
    n_nw = np.sum(nw_quadrant)
    n_sw = np.sum(sw_quadrant)

    # Convert to percentages
    pct_ne = 100 * n_ne / n_iterations
    pct_se = 100 * n_se / n_iterations
    pct_nw = 100 * n_nw / n_iterations
    pct_sw = 100 * n_sw / n_iterations

    print(f"Total PSA iterations: {n_iterations}")

    print(
         f"NE - More effective, more costly: "
         f"{n_ne} ({pct_ne:.1f}%)"
    )

    print(
         f"SE - More effective, less costly: "
         f"{n_se} ({pct_se:.1f}%)"
    )

    print(
         f"NW - Less effective, more costly: "
         f"{n_nw} ({pct_nw:.1f}%)"
    )

    print(
        f"SW - Less effective, less costly: "
        f"{n_sw} ({pct_sw:.1f}%)"
    )

    print(
        f"Total across quadrants: "
        f"{pct_ne + pct_se + pct_nw + pct_sw:.1f}%"
    )

    # Range of incremental costs and net effects
    inc_cost_min, inc_cost_max = np.min(Incremental_costs_sample), np.max(Incremental_costs_sample)
    inc_eff_min, inc_eff_max = np.min(Incremental_net_effects_sample), np.max(Incremental_net_effects_sample)

    print(f"Incremental costs ranged from £{inc_cost_min:,.0f} to £{inc_cost_max:,.0f}")
    print(f"Incremental net effects ranged from {inc_eff_min:.3f} to {inc_eff_max:.3f}")

    # COST-EFFECTIVENESS PLANE

    from matplotlib import pyplot as plt

    # limits for symmetry
    max_effect = max(abs(min(Incremental_net_effects_sample)), abs(max(Incremental_net_effects_sample)))
    max_cost = max(abs(min(Incremental_costs_sample)), abs(max(Incremental_costs_sample)))

    figce = plt.figure(figsize=(7, 6))
    plt.axhline(0, color='black', linestyle='-')
    plt.axvline(0, color='black', linestyle='-')

    # Mean point
    mean_incremental_cost = Incremental_cost
    mean_incremental_effect = Incremental_net_effect

    # Plotting data points
    plt.scatter(Incremental_net_effects_sample, Incremental_costs_sample, color='blue', alpha=0.25, s=8, label='PSA Samples')

    # NICE WTP line
    wtp = 30000
    x_vals = np.linspace(0, max_effect * 1.1, 100)
    plt.plot(x_vals, wtp * x_vals, '--', color='green', label=f'NICE WTP = £{wtp:,}')

    import matplotlib.ticker as ticker

    # Y-axis to show full numbers
    plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, _: f'{int(x):,}'))

    # Mean incremental point
    plt.scatter(mean_incremental_effect, mean_incremental_cost, color='black', marker='x', s=100, label='Mean')

    # ICER line
    icer_slope = mean_incremental_cost / mean_incremental_effect if mean_incremental_effect != 0 else 0
    plt.plot([0, max_effect * 1.1], [0, icer_slope * max_effect * 1.1], color='red', linestyle='--', label='ICER')

    plt.xlim(-max_effect * 1.1, max_effect * 1.1)
    plt.ylim(-max_cost * 0.5, max_cost * 1.1)

    plt.xlabel("Incremental Net Effect")
    plt.ylabel("Incremental Cost (£)")
    plt.title("Cost-Effectiveness Plane")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    figce.savefig('/content/drive/My Drive/Colab Notebooks/Cost effectiveness plane.png', dpi=300, bbox_inches='tight')
    plt.show()

    # COST-ACCEPTABILITY CURVE

    import matplotlib.ticker as mtick

    # Initialize parameters
    wtp_step = 5000
    wtp_max_limit = 300000  # Limit
    wtp_current = 0

    wtp_range = []
    ce_probabilities = []

    # Loop until CE probability reaches 100%
    while wtp_current <= wtp_max_limit:
       # Calculating Net Monetary Benefit for the WTP
       nmb = [wtp_current * effect - cost for effect, cost in zip(Incremental_net_effects_sample, Incremental_costs_sample)]

       # Proportion of PSA samples with Net monetary benefit > 0
       ce_prob = np.mean([1 if x > 0 else 0 for x in nmb])

       wtp_range.append(wtp_current)
       ce_probabilities.append(ce_prob)

       # Stop if Cost-effectiveness probability is 100%
       if ce_prob >= 0.999:
          break

       wtp_current += wtp_step

    # Plotting curve
    figca = plt.figure(figsize=(7, 5))
    plt.plot(wtp_range, ce_probabilities, color='darkgreen')

    # NICE WTP line
    plt.axvline(30000, color='slategray', linestyle='--', label='NICE WTP = £30,000')

    plt.xlabel("Willingness to Pay (£)")
    plt.ylabel("Probability Cost-Effective")
    plt.title("Cost-Effectiveness Acceptability Curve")

    plt.xlim(0, 300000)  # x-axis limit

    tick_interval = 30000
    xticks = np.arange(0, 300001, tick_interval)
    plt.xticks(xticks, rotation=45)

    plt.ylim(0, 1)  # y-axis limit

    plt.gca().yaxis.set_major_formatter(mtick.PercentFormatter(1.0))

    plt.grid(True)
    plt.tight_layout()
    plt.legend()
    figca.savefig('/content/drive/My Drive/Colab Notebooks/Cost acceptability curve.png', dpi=300, bbox_inches='tight')
    plt.show()




# -------------------------------------------------------
# 7. DETERMINISTIC ANALYSES
# -------------------------------------------------------
# 7.1 DEFINE DETERMINISTIC
# -----------------------------

# Decision tree
dt_base_draws = {
    "COST_FHQS_TEXT": PARAMS["COST_FHQS_TEXT"].mean(),
    "COST_UPTAKE": PARAMS["COST_UPTAKE"].mean(),
    "COST_HIGH_RISK": PARAMS["COST_HIGH_RISK"],
    "COST_REFERRAL": PARAMS["COST_REFERRAL"],

    "P_FHQS_UPTAKE": PARAMS["P_FHQS_UPTAKE"].mean(),

    "P_RISK_SPLIT_FHQS": PARAMS[
        "P_RISK_SPLIT_FHQS"
    ].mean(),

    "IMD_REFERRAL": {
        imd: PARAMS["IMD_REFERRAL"][imd].mean()
        for imd in imd_levels
    },

    "P_RISK_SPLIT_REFERRAL": PARAMS[
        "P_RISK_SPLIT_REFERRAL"
    ].mean(),
}
# MODEL 1
referral_base_draws = {
    imd: IMD_referral_probs[imd].mean()
    for imd in imd_levels
}

risk_split_base = P_RISK_SPLIT_REFERRAL.mean()

symptomatic_base_draws = {
    age: symptomatic_detection_probs[age].mean()
    for age in AGE_RANGE
}

symptomatic_cost_base = symptomatic_detection_cost.mean()
# MODEL 2
sens_base = SENS_MEAN
spec_base = SPEC_MEAN
uptake_base = UP_MEAN

model2_base_costs = {
    "Healthy_TN": cost_params["Healthy_TN"],
    "Healthy_FP": cost_params["Healthy_FP"].mean(),
    "Cancer_Undetected": cost_params["Cancer_Undetected"],
    "Cancer_Detected": cost_params["Cancer_Detected"].mean()
}


# 7.2 SIMULATE DECISION TREE
# ------------------------------------------

def run_decision_tree_deterministic(
    cohort,
    dt_draws,
    fhqs_uptake_by_imd=None,
):

    cohort = cohort.copy()

    # Deterministic model requires population weights
    if "n" not in cohort.columns:
        raise ValueError(
            "Deterministic cohort is missing 'n'. "
            "The population weights must be supplied before "
            "run_decision_tree_deterministic()."
        )

    indiv_control = []
    indiv_intervention = []

    total_control = {
        "cost": 0.0,
        "Unclassified": 0.0,
        "Low_risk": 0.0,
        "Moderate_risk": 0.0,
        "High_risk": 0.0
    }

    total_intervention = {
        "cost": 0.0,
        "Unclassified": 0.0,
        "Low_risk": 0.0,
        "Moderate_risk": 0.0,
        "High_risk": 0.0
    }

    for _, person in cohort.iterrows():

        imd = int(person["imd"])
        start_age = int(person["start_age"])
        n = float(person["n"])

        # CONTROL

        p_referral = dt_draws["IMD_REFERRAL"][imd]
        referral_split = dt_draws["P_RISK_SPLIT_REFERRAL"]

        p_low = p_referral * referral_split[0]
        p_mod = p_referral * referral_split[1]
        p_high = p_referral * referral_split[2]
        p_unclassified = 1 - p_referral

        cost_control = (
              p_referral * dt_draws["COST_REFERRAL"]
              + p_high * dt_draws["COST_HIGH_RISK"]
        )

        total_control["cost"] += n * cost_control
        total_control["Unclassified"] += n * p_unclassified
        total_control["Low_risk"] += n * p_low
        total_control["Moderate_risk"] += n * p_mod
        total_control["High_risk"] += n * p_high

        # cohort for downstream models
        indiv_control.extend([
             {
                "start_age": start_age,
                "imd": imd,
                "risk": "unclassified",
                "n": n * p_unclassified
             },
             {
                "start_age": start_age,
                "imd": imd,
                "risk": "low_risk",
                "n": n * p_low
             },
             {
                "start_age": start_age,
                "imd": imd,
                "risk": "moderate_risk",
                "n": n * p_mod
             },
             {
                "start_age": start_age,
                "imd": imd,
                "risk": "high_risk",
                "n": n * p_high
             }
        ])

        # INTERVENTION

        if fhqs_uptake_by_imd is not None:
           # DCEA: IMD-specific FHQS uptake
           p_uptake = fhqs_uptake_by_imd[imd]
        else:
           # OWSA: overall FHQS uptake
           p_uptake = dt_draws["P_FHQS_UPTAKE"]

        fhqs_split = dt_draws["P_RISK_SPLIT_FHQS"]

        p_low_fhqs = p_uptake * fhqs_split[0]
        p_mod_fhqs = p_uptake * fhqs_split[1]
        p_high_fhqs = p_uptake * fhqs_split[2]

        p_unclassified_fhqs = 1 - p_uptake

        cost_intervention = (
            dt_draws["COST_FHQS_TEXT"]
            + p_uptake * dt_draws["COST_UPTAKE"]
            + p_high_fhqs * dt_draws["COST_HIGH_RISK"]
        )

        total_intervention["cost"] += n * cost_intervention
        total_intervention["Unclassified"] += n * p_unclassified_fhqs
        total_intervention["Low_risk"] += n * p_low_fhqs
        total_intervention["Moderate_risk"] += n * p_mod_fhqs
        total_intervention["High_risk"] += n * p_high_fhqs

        indiv_intervention.extend([
            {
               "start_age": start_age,
               "imd": imd,
               "risk": "unclassified",
               "n": n * p_unclassified_fhqs
            },
            {
               "start_age": start_age,
               "imd": imd,
               "risk": "low_risk",
               "n": n * p_low_fhqs
            },
            {
               "start_age": start_age,
               "imd": imd,
               "risk": "moderate_risk",
               "n": n * p_mod_fhqs
            },
            {
               "start_age": start_age,
               "imd": imd,
               "risk": "high_risk",
               "n": n * p_high_fhqs
            }
        ])

    return (
        total_control,
        total_intervention,
        pd.DataFrame(indiv_control),
        pd.DataFrame(indiv_intervention)
    )

# 7.3 SIMULATE FULL MODEL
# -------------------------------------------
def run_full_model_deterministic(
    cohort,
    dt_draws,
    referral_draws,
    risk_split_draw,
    symptomatic_detection_draws,
    symptomatic_cost_draw,
    sens_draw,
    spec_draw,
    uptake_draw,
    model2_cost_draws,
    fhqs_uptake_by_imd
):

    cohort = cohort.copy()

    (
       total_control,
       total_interv,
       indiv_control_df,
       indiv_interv_df
    ) = run_decision_tree_deterministic(
       cohort=cohort,
       dt_draws=dt_draws,
       fhqs_uptake_by_imd=fhqs_uptake_by_imd
    )
     # MARKOV MODEL 1 COHORTS
    control_model1_cohort = (
        indiv_control_df[
            indiv_control_df["risk"].isin(
                ["unclassified", "low_risk"]
            )
        ]
        .copy()
    )

    interv_model1_cohort = (
        indiv_interv_df[
            indiv_interv_df["risk"].isin(
                ["unclassified", "low_risk"]
            )
        ]
        .copy()
    )
    # MARKOV MODEL 1
    (
        control_model1_results,
        control_model1_identified,
        control_model1_cost,
        control_model1_symptomatic
    ) = run_markov_model(
        cohort=control_model1_cohort,
        arm="control",
        referral_draws=referral_draws,
        risk_split_draw=risk_split_draw,
        symptomatic_detection_draws=symptomatic_detection_draws,
        symptomatic_cost_draw=symptomatic_cost_draw
    )

    (
        interv_model1_results,
        interv_model1_identified,
        interv_model1_cost,
        interv_model1_symptomatic
    ) = run_markov_model(
        cohort=interv_model1_cohort,
        arm="intervention",
        referral_draws=referral_draws,
        risk_split_draw=risk_split_draw,
        symptomatic_detection_draws=symptomatic_detection_draws,
        symptomatic_cost_draw=symptomatic_cost_draw
    )
    # Aggregrate results by imd for DCEA analysis
    # CONTROL
    control_model1_cost_by_imd = (
      control_model1_results
      .groupby("imd")["cost"]
      .sum()
    )

    control_model1_symptomatic_by_imd = (
      control_model1_results
      .groupby("imd")["symptomatic_detected"]
      .sum()
    )

    # INTERVENTION
    interv_model1_cost_by_imd = (
       interv_model1_results
       .groupby("imd")["cost"]
       .sum()
    )

    interv_model1_symptomatic_by_imd = (
       interv_model1_results
       .groupby("imd")["symptomatic_detected"]
       .sum()
    )
    # MARKOV MODEL 2 ENTRANTS
    # Direct from decision tree
    control_original_model2 = (
        indiv_control_df[
            indiv_control_df["risk"].isin(
                ["moderate_risk", "high_risk"]
            )
        ]
        .groupby(["risk", "start_age", "imd"])["n"]
        .sum()
        .reset_index()
    )

    interv_original_model2 = (
        indiv_interv_df[
            indiv_interv_df["risk"].isin(
                ["moderate_risk", "high_risk"]
            )
        ]
        .groupby(["risk", "start_age", "imd"])["n"]
        .sum()
        .reset_index()
    )


    # From Model 1
    control_new_model2 = (
        control_model1_identified[
            control_model1_identified["risk"].isin(
                ["moderate_risk", "high_risk"]
            )
        ]
        .rename(columns={"age": "start_age"})
        .groupby(["risk", "start_age", "imd"])["n"]
        .sum()
        .reset_index()
    )

    interv_new_model2 = (
        interv_model1_identified[
            interv_model1_identified["risk"].isin(
                ["moderate_risk", "high_risk"]
            )
        ]
        .rename(columns={"age": "start_age"})
        .groupby(["risk", "start_age", "imd"])["n"]
        .sum()
        .reset_index()
    )


    # Combine Model 2 cohorts
    control_model2_cohort = pd.concat(
        [
            control_original_model2,
            control_new_model2
        ],
        ignore_index=True
    )

    interv_model2_cohort = pd.concat(
        [
            interv_original_model2,
            interv_new_model2
        ],
        ignore_index=True
    )
    # MARKOV MODEL 2 MATRICES
    model2_matrices = build_matrices(
        prevalence_df,
        sens_draw,
        spec_draw,
        uptake_draw
    )

    # MARKOV MODEL 2
    control_model2_cost, control_model2_effect = run_model2(
      control_model2_cohort,
      model2_cost_draws,
      model2_matrices
    )

    control_model2_cost_by_imd = (
      control_model2_cost
      .groupby("imd")["cost"]
      .sum()
    )

    control_model2_screen_detected_by_imd = (
      control_model2_effect
      .groupby("imd")["detected"]
      .sum()
    )

    interv_model2_cost, interv_model2_effect = run_model2(
      interv_model2_cohort,
      model2_cost_draws,
      model2_matrices
    )

    interv_model2_cost_by_imd = (
      interv_model2_cost
      .groupby("imd")["cost"]
      .sum()
    )

    interv_model2_screen_detected_by_imd = (
      interv_model2_effect
      .groupby("imd")["detected"]
      .sum()
    )

     # TOTAL COSTS

    control_total_cost = (
        total_control["cost"]
        + control_model1_cost
        + control_model2_cost["cost"].sum()
    )

    interv_total_cost = (
        total_interv["cost"]
        + interv_model1_cost
        + interv_model2_cost["cost"].sum()
    )

    # DECISION-TREE COSTS BY IMD

    decision_tree_cost_by_imd_control = {}
    decision_tree_cost_by_imd_intervention = {}

    for _, person in cohort.iterrows():

        imd = int(person["imd"])
        n = float(person["n"])

        # ----------------------------
        # CONTROL
        # ----------------------------

        p_referral = dt_draws["IMD_REFERRAL"][imd]
        referral_split = dt_draws["P_RISK_SPLIT_REFERRAL"]

        p_low = p_referral * referral_split[0]
        p_mod = p_referral * referral_split[1]
        p_high = p_referral * referral_split[2]

        cost_control_per_person = (
            p_referral * dt_draws["COST_REFERRAL"]
            + p_high * dt_draws["COST_HIGH_RISK"]
        )

        decision_tree_cost_by_imd_control[imd] = (
            decision_tree_cost_by_imd_control.get(imd, 0.0)
            + cost_control_per_person * n
        )

        # ----------------------------
        # INTERVENTION
        # ----------------------------

        # INTERVENTION

        if fhqs_uptake_by_imd is not None:
           # DCEA: IMD-specific FHQS uptake
           p_uptake = fhqs_uptake_by_imd[imd]
        else:
           # OWSA: overall FHQS uptake
           p_uptake = dt_draws["P_FHQS_UPTAKE"]

        fhqs_split = dt_draws["P_RISK_SPLIT_FHQS"]

        p_low_fhqs = p_uptake * fhqs_split[0]
        p_mod_fhqs = p_uptake * fhqs_split[1]
        p_high_fhqs = p_uptake * fhqs_split[2]

        cost_intervention_per_person = (
            dt_draws["COST_FHQS_TEXT"]
            + p_uptake * dt_draws["COST_UPTAKE"]
            + p_high_fhqs * dt_draws["COST_HIGH_RISK"]
        )

        decision_tree_cost_by_imd_intervention[imd] = (
            decision_tree_cost_by_imd_intervention.get(imd, 0.0)
            + cost_intervention_per_person * n
        )

    decision_tree_cost_by_imd_control = pd.Series(
        decision_tree_cost_by_imd_control
    )

    decision_tree_cost_by_imd_intervention = pd.Series(
        decision_tree_cost_by_imd_intervention
    )

    # IMD-LEVEL DCEA TABLE

    imd_values = sorted(cohort["imd"].unique())

    dcea_results = pd.DataFrame({
        "imd": imd_values
    })

    # POPULATION

    population_by_imd = (
        cohort
        .groupby("imd")["n"]
        .sum()
    )

    dcea_results["population"] = (
        dcea_results["imd"]
        .map(population_by_imd)
        .fillna(0)
    )


    # COSTS

    # CONTROL

    dcea_results["control_cost"] = (
        dcea_results["imd"]
        .map(decision_tree_cost_by_imd_control)
        .fillna(0)
        +
        dcea_results["imd"]
        .map(control_model1_cost_by_imd)
        .fillna(0)
        +
        dcea_results["imd"]
        .map(control_model2_cost_by_imd)
        .fillna(0)
    )

    # INTERVENTION

    dcea_results["intervention_cost"] = (
        dcea_results["imd"]
        .map(decision_tree_cost_by_imd_intervention)
        .fillna(0)
        +
        dcea_results["imd"]
        .map(interv_model1_cost_by_imd)
        .fillna(0)
        +
        dcea_results["imd"]
        .map(interv_model2_cost_by_imd)
        .fillna(0)
    )

    # INCREMENTAL COST

    dcea_results["incremental_cost"] = (
        dcea_results["intervention_cost"]
        - dcea_results["control_cost"]
    )

    # COST PER WOMAN

    dcea_results["control_cost_per_woman"] = (
        dcea_results["control_cost"]
        / dcea_results["population"]
    )

    dcea_results["intervention_cost_per_woman"] = (
        dcea_results["intervention_cost"]
        / dcea_results["population"]
    )

    dcea_results["incremental_cost_per_woman"] = (
        dcea_results["incremental_cost"]
        / dcea_results["population"]
    )


    # SYMPTOMATIC DETECTIONS

    dcea_results["control_symptomatic"] = (
        dcea_results["imd"]
        .map(control_model1_symptomatic_by_imd)
        .fillna(0)
    )

    dcea_results["intervention_symptomatic"] = (
        dcea_results["imd"]
        .map(interv_model1_symptomatic_by_imd)
        .fillna(0)
    )

    dcea_results["incremental_symptomatic"] = (
        dcea_results["intervention_symptomatic"]
        - dcea_results["control_symptomatic"]
    )

    # SCREEN-DETECTED CANCERS

    dcea_results["control_screen_detected"] = (
        dcea_results["imd"]
        .map(control_model2_screen_detected_by_imd)
        .fillna(0)
    )

    dcea_results["intervention_screen_detected"] = (
        dcea_results["imd"]
        .map(interv_model2_screen_detected_by_imd)
        .fillna(0)
    )

    dcea_results["incremental_screen_detected"] = (
        dcea_results["intervention_screen_detected"]
        - dcea_results["control_screen_detected"]
    )

    # NET HEALTH EFFECT

    dcea_results["control_net_effect"] = (
        dcea_results["control_screen_detected"]
        - dcea_results["control_symptomatic"]
    )

    dcea_results["intervention_net_effect"] = (
        dcea_results["intervention_screen_detected"]
        - dcea_results["intervention_symptomatic"]
    )

    dcea_results["incremental_net_effect"] = (
        dcea_results["intervention_net_effect"]
        - dcea_results["control_net_effect"]
    )

    # FINAL RESULTS

    return {
        # Overall results used by PSA / OWSA
        "control_cost": control_total_cost,
        "interv_cost": interv_total_cost,

        "control_symptomatic": control_model1_symptomatic,
        "interv_symptomatic": interv_model1_symptomatic,

        "control_screen_detected": (
            control_model2_effect["detected"].sum()
        ),

        "interv_screen_detected": (
            interv_model2_effect["detected"].sum()
        ),

        # IMD-level DCEA results
        "dcea_results": dcea_results
    }

print("Cohort rows:", len(cohort))
print("Cohort population:", cohort["n"].sum())
print(cohort[["start_age", "imd", "n"]].head(10))


# 7.4 ONE-WAY SENSITIVITY ANALYSIS
# ----------------------------------------
# 7.4.1 DEFINE SIMULATION FUNCTION
# ----------------------------------------
def run_owsa(
    parameter,
    values
):

    results = []

    for value in values:

        # Starting with base-case values
        scenario_cohort = cohort.copy()

        dt_draws = dt_base_draws.copy()

        referral_draws = referral_base_draws.copy()

        risk_split_draw = risk_split_base

        symptomatic_detection_draws = (
            symptomatic_base_draws.copy()
        )

        symptomatic_cost_draw = symptomatic_cost_base

        sens_draw = sens_base
        spec_draw = spec_base
        uptake_draw = uptake_base

        model2_cost_draws = model2_base_costs.copy()

        # Changing one parameter
        if parameter == "P_FHQS_UPTAKE":

            dt_draws["P_FHQS_UPTAKE"] = value

        elif parameter == "Sample size (n)":
            scenario_cohort = cohort.copy()
            scenario_cohort["n"] = (
                 scenario_cohort["n"] / scenario_cohort["n"].sum()
            ) * value

        elif parameter == "BC risk after FHQS (Dirichlet)":

            dt_draws["P_RISK_SPLIT_FHQS"] = np.array(value) / np.sum(value)


        elif parameter == "COST_FHQS_TEXT":

            dt_draws["COST_FHQS_TEXT"] = value


        elif parameter == "COST_SYMPTOM":

            symptomatic_cost_draw = value


        elif parameter == "SENSITIVITY":

            sens_draw = value


        elif parameter == "SPECIFICITY":

            spec_draw = value


        elif parameter == "SCREENING_UPTAKE":

            uptake_draw = value


        elif parameter == "CANCER_DETECTED_COST":

            model2_cost_draws[
                "Cancer_Detected"
            ] = value


        elif parameter == "FALSE_POSITIVE_COST":

            model2_cost_draws[
                "Healthy_FP"
            ] = value


        else:

            raise ValueError(
                f"Unsupported OWSA parameter: {parameter}"
            )

        # Running complete model
        result = run_full_model_deterministic(

            cohort=scenario_cohort if parameter == "Sample size (n)" else cohort,

            dt_draws=dt_draws,

            referral_draws=referral_draws,
            risk_split_draw=risk_split_draw,
            symptomatic_detection_draws=
                symptomatic_detection_draws,
            symptomatic_cost_draw=
                symptomatic_cost_draw,

            sens_draw=sens_draw,
            spec_draw=spec_draw,
            uptake_draw=uptake_draw,

            model2_cost_draws=model2_cost_draws,
            fhqs_uptake_by_imd=None
        )
        # Incremental outcomes
        incremental_cost_OWSA = (
            result["interv_cost"]
            - result["control_cost"]
        )

        Net_effect_control_OWSA = (
            result["control_screen_detected"]
            - result["control_symptomatic"]
        )

        Net_effect_interv_OWSA = (
            result["interv_screen_detected"]
            - result["interv_symptomatic"]
        )

        incremental_net_effect_OWSA = (Net_effect_interv_OWSA - Net_effect_control_OWSA)

        if incremental_net_effect_OWSA != 0:

            icer = (
                incremental_cost_OWSA
                / incremental_net_effect_OWSA
            )
        else:
            icer = np.nan
        # Store results
        results.append({

            "parameter": parameter,

            "value": value,

            "control_cost":
                result["control_cost"],

            "intervention_cost":
                result["interv_cost"],

            "incremental_cost":
                incremental_cost_OWSA,

            "control_symptomatic":
                result["control_symptomatic"],

            "intervention_symptomatic":
                result["interv_symptomatic"],

            "control_screen_detected":
                result["control_screen_detected"],

            "intervention_screen_detected":
                result["interv_screen_detected"],

            "control_net_effect":
                Net_effect_control_OWSA,

            "intervention_net_effect":
                Net_effect_interv_OWSA,

            "incremental_net_effect":
                incremental_net_effect_OWSA,

            "ICER":
                icer
        })
    return pd.DataFrame(results)



# 7.4.2 OWSA PARAMETER RANGES
# ------------------------------------

owsa_results = {}

# FHQS uptake
owsa_results["P_FHQS_UPTAKE"] = run_owsa(
    "P_FHQS_UPTAKE",
    np.linspace(0.05, 0.25, 5)
)
# Cost of FHQS set up
owsa_results["COST_FHQS_TEXT"] = run_owsa(
    "COST_FHQS_TEXT",
    np.linspace(10, 50, 5)
)

# BC risk after FHQS (Dirichlet)
owsa_results["BC risk after FHQS (Dirichlet)"] = run_owsa(
    "BC risk after FHQS (Dirichlet)",
    [
        (80.0, 18.23, 1.77),
        (83.0, 15.5, 1.5),
        (87.0, 11.85, 1.15),
        (91.0, 8.2, 0.8),
        (95.0, 4.56, 0.44),
    ]
)

# Sample size variation
owsa_results["Sample size (n)"] = run_owsa(
    "Sample size (n)",
    [20000, 40000, 60000, 80000, 100000]
)

# Symptomatic detection cost
owsa_results["COST_SYMPTOM"] = run_owsa(
    "COST_SYMPTOM",
    np.linspace(800, 1500, 5)
)

# Mammography sensitivity
owsa_results["SENSITIVITY"] = run_owsa(
    "SENSITIVITY",
    np.linspace(0.68, 0.88, 5)
)


# Mammography specificity
owsa_results["SPECIFICITY"] = run_owsa(
    "SPECIFICITY",
    np.linspace(0.85, 0.95, 5)
)


# Screening uptake
owsa_results["SCREENING_UPTAKE"] = run_owsa(
    "SCREENING_UPTAKE",
    np.linspace(0.65, 0.95, 5)
)


# Screening detection cost
owsa_results["CANCER_DETECTED_COST"] = run_owsa(
    "CANCER_DETECTED_COST",
    np.linspace(600, 1200, 5)
)


# False-positive cost
owsa_results["FALSE_POSITIVE_COST"] = run_owsa(
    "FALSE_POSITIVE_COST",
    np.linspace(400, 700, 5)
)

# 7.4.3 OWSA SUMMARY TABLE
# ----------------------------------------

owsa_summary_rows = []

for parameter, df in owsa_results.items():

    for _, r in df.iterrows():

        # Control arm
        owsa_summary_rows.append({
            "Parameter": parameter,
            "Parameter value": r["value"],
            "Arm": "Control",
            "Total cost (£)": r["control_cost"],
            "Screen-detected cases": r["control_screen_detected"],
            "Symptomatic-detected cases": r["control_symptomatic"],
            "Net-effect": r["control_net_effect"],
            "ICER (£ per net detection)": np.nan
        })

        owsa_summary_rows.append({
            "Parameter": parameter,
            "Parameter value": r["value"],
            "Arm": "Intervention",
            "Total cost (£)": r["intervention_cost"],
            "Screen-detected cases": r["intervention_screen_detected"],
            "Symptomatic-detected cases": r["intervention_symptomatic"],
            "Net-effect": r["intervention_net_effect"],
            "ICER (£ per net detection)": r["ICER"]
        })

owsa_summary = pd.DataFrame(owsa_summary_rows)

# Round for presentation
numeric_columns = owsa_summary.select_dtypes(
    include="number"
).columns

owsa_summary[numeric_columns] = (
    owsa_summary[numeric_columns].round(2)
)

print(owsa_summary.to_string(index=False))



# 7.4.4 TORNADO PLOT FOR OWSA
# ------------------------------------

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Base-case ICER
base_case_result = run_full_model_deterministic(
    cohort=cohort,
    dt_draws=dt_base_draws,
    referral_draws=referral_base_draws,
    risk_split_draw=risk_split_base,
    symptomatic_detection_draws=symptomatic_base_draws,
    symptomatic_cost_draw=symptomatic_cost_base,
    sens_draw=sens_base,
    spec_draw=spec_base,
    uptake_draw=uptake_base,
    model2_cost_draws=model2_base_costs,
    fhqs_uptake_by_imd=None
)

base_incremental_cost = (
    base_case_result["interv_cost"]
    - base_case_result["control_cost"]
)

base_incremental_effect = ((base_case_result["interv_screen_detected"] - base_case_result["interv_symptomatic"]) - (base_case_result["control_screen_detected"] - base_case_result["control_symptomatic"]))

base_case_icer = (
    base_incremental_cost / base_incremental_effect
)

# Convert OWSA results into the format needed for plotting
data = {}

for param, df in owsa_results.items():

    data[param] = {
        "outcomes": df["ICER"].tolist(),
        "values": df["value"].tolist()
    }


# Sort parameters by largest ICER range, descending
sorted_params = sorted(
    data.items(),
    key=lambda x: max(x[1]["outcomes"]) - min(x[1]["outcomes"]),
    reverse=True
)


# Create figure
fig, ax = plt.subplots(figsize=(12, 8))


# Base-case ICER
base_case = base_case_icer


# Add base-case vertical line
base_line = ax.axvline(
    base_case,
    linestyle="--",
    color="gray",
    linewidth=1.5
)


# Legend
ax.legend(
    [base_line],
    [f"Base case: £{base_case:,.0f}"],
    loc="lower right",
    frameon=True,
    edgecolor="black"
)


ytick_labels = []

parameter_labels = {
    "P_FHQS_UPTAKE": "FHQS uptake probability",
    "P_RISK_SPLIT_FHQS": "BC risk after FHQS (Dirichlet)",
    "BC risk after FHQS (Dirichlet)": "BC risk after FHQS (Dirichlet)",
    "Sample size (n)": "Sample size",
    "COST_FHQS_TEXT": "FHQS implementation cost per person",
    "COST_SYMPTOM": "Symptomatic detection cost per person",
    "SENSITIVITY": "Mammography sensitivity",
    "SPECIFICITY": "Mammography specificity",
    "SCREENING_UPTAKE": "Screening uptake probability",
    "CANCER_DETECTED_COST": "Cancer detection cost per person",
    "FALSE_POSITIVE_COST": "False positive cost per person"
}


# Plot each parameter
for i, (param, vals) in enumerate(sorted_params):

    # Remove missing or infinite ICERs before plotting
    valid_pairs = [
        (outcome, value)
        for outcome, value in zip(vals["outcomes"], vals["values"])
        if pd.notna(outcome) and np.isfinite(outcome)
    ]

    if not valid_pairs:
        continue

    outcomes, values = zip(*valid_pairs)
    y_vals = [i] * len(outcomes)

    ax.plot(
        outcomes,
        y_vals,
        marker="o",
        linestyle="-",
        label="_nolegend_"
    )

    # Parameter value range
    first_value = values[0]
    last_value = values[-1]

    if param in [
        "P_FHQS_UPTAKE",
        "SENSITIVITY",
        "SPECIFICITY",
        "SCREENING_UPTAKE"
    ]:
        label_summary = (
            f"{first_value:.0%} → {last_value:.0%}"
        )

    elif param in [
        "COST_FHQS_TEXT",
        "COST_SYMPTOM",
        "CANCER_DETECTED_COST",
        "FALSE_POSITIVE_COST"
    ]:
        label_summary = (
            f"£{first_value:,.0f} → £{last_value:,.0f}"
        )

    elif param in [
        "BC risk after FHQS (Dirichlet)",
        "P_RISK_SPLIT_FHQS"
    ]:
        # Values are three-component Dirichlet parameter sets
        label_summary = (
            f"{tuple(first_value)} → {tuple(last_value)}"
        )
    elif param == "Sample size (n)":
        label_summary = f"{first_value:,.0f} → {last_value:,.0f}"

    else:
        label_summary = (
            f"{first_value:,.2f} → {last_value:,.2f}"
        )

    display_name = parameter_labels.get(param, param)

    ytick_labels.append(
        f"{display_name}\n[{label_summary}]"
    )


# Set y-axis ticks to match the plotted parameters
# (handles any parameters skipped because all ICERs were invalid)
plotted_params = [
    (param, vals)
    for param, vals in sorted_params
    if any(
        pd.notna(outcome) and np.isfinite(outcome)
        for outcome in vals["outcomes"]
    )
]

ax.set_yticks(range(len(plotted_params)))
ax.set_yticklabels(ytick_labels, fontsize=9)


# Labels
ax.set_xlabel("ICER (£)")
ax.set_title("One-Way Sensitivity Analysis Funnel Plot")


# Grid
ax.grid(True, axis="x")


# Put largest range at top
ax.invert_yaxis()


plt.tight_layout()


# Save
fig.savefig(
    "/content/drive/My Drive/Colab Notebooks/OWSA_funnel_plot.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()

# -------------------------------------------------
# 8. DISTRIBUTIONAL COST-EFFECTIVENESS ANALYSES
# -------------------------------------------------

# 8.1 FHQS uptake probabilities by IMD decile
# -----------------------------------------------

fhqs_uptake_by_imd = {
    1: 0.07864,
    2: 0.0942,
    3: 0.10549,
    4: 0.11659,
    5: 0.13111,
    6: 0.14577,
    7: 0.15741,
    8: 0.16906,
    9: 0.18358,
    10: 0.19536
}

# 8.2 RUN FULL MODEL FOR DCEA
# -----------------------------------------

dcea_output = run_full_model_deterministic(
    cohort=cohort,
    dt_draws=dt_base_draws,
    referral_draws=referral_base_draws,
    risk_split_draw=risk_split_base,
    symptomatic_detection_draws=symptomatic_base_draws,
    symptomatic_cost_draw=symptomatic_cost_base,
    sens_draw=sens_base,
    spec_draw=spec_base,
    uptake_draw=uptake_base,
    model2_cost_draws=model2_base_costs,
    fhqs_uptake_by_imd=fhqs_uptake_by_imd
)

dcea_results = dcea_output["dcea_results"]

dcea_df = dcea_output["dcea_results"].copy()

# Calculate ICER by IMD decile

dcea_df["ICER"] = (
    dcea_df["incremental_cost"] /
    dcea_df["incremental_net_effect"]
)

icer_by_imd = dcea_df[
    [
        "imd",
        "population",
        "incremental_cost",
        "incremental_net_effect",
        "ICER"
    ]
].copy()

display(icer_by_imd.round(2))


# 8.3 OUTCOME DISTRIBUTIONS BY IMD PRE AND POST INTERVENTION
# ----------------------------------------------------------
# COSTS
cost_plot = dcea_df[
    [
        "imd",
        "control_cost_per_woman",
        "intervention_cost_per_woman",
        "incremental_cost_per_woman"
    ]
].copy()

cost_plot = cost_plot.rename(columns={
    "imd": "IMD",
    "control_cost_per_woman": "NO_TEXT",
    "intervention_cost_per_woman": "SEND_TEXT",
    "incremental_cost_per_woman": "Incremental"
})

x = np.arange(len(cost_plot["IMD"]))
width = 0.25

# Create figure and axes
fig, ax = plt.subplots(figsize=(12, 6))

ax.bar(
    x - width,
    cost_plot["NO_TEXT"],
    width,
    label="Before intervention"
)

ax.bar(
    x,
    cost_plot["SEND_TEXT"],
    width,
    label="After intervention"
)

ax.bar(
    x + width,
    cost_plot["Incremental"],
    width,
    label="Incremental cost"
)

ax.axhline(0, linewidth=0.8)

ax.set_xlabel("IMD decile")
ax.set_ylabel("Cost per woman (£)")
ax.set_title("Distribution of cost per woman by IMD decile")

ax.set_xticks(x)
ax.set_xticklabels(cost_plot["IMD"])

# Format y-axis as £ with commas
ax.yaxis.set_major_formatter(
    ticker.FuncFormatter(lambda x, pos: f"£{x:,.0f}")
)

ax.legend()

plt.tight_layout()

# Save the actual Matplotlib figure
fig.savefig(
    "/content/drive/My Drive/Colab Notebooks/DCEA_cost_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# NET EFFECT

net_effect_plot = dcea_df[
    [
        "imd",
        "control_net_effect",
        "intervention_net_effect",
        "incremental_net_effect"
    ]
].copy()

net_effect_plot = net_effect_plot.rename(columns={
    "imd": "IMD",
    "control_net_effect": "NO_TEXT",
    "intervention_net_effect": "SEND_TEXT",
    "incremental_net_effect": "Incremental"
})

x = np.arange(len(net_effect_plot["IMD"]))
width = 0.25

fig, ax = plt.subplots(figsize=(12, 6))

ax.bar(
    x - width,
    net_effect_plot["NO_TEXT"],
    width,
    label="Before intervention"
)

ax.bar(
    x,
    net_effect_plot["SEND_TEXT"],
    width,
    label="After intervention"
)

ax.bar(
    x + width,
    net_effect_plot["Incremental"],
    width,
    label="Incremental net effect"
)

ax.axhline(0, linewidth=0.8)

ax.set_xlabel("IMD decile")
ax.set_ylabel("Net effect")
ax.set_title("Distribution of net effect by IMD decile")

ax.set_xticks(x)
ax.set_xticklabels(net_effect_plot["IMD"])
ax.legend()

plt.tight_layout()

fig.savefig(
    "/content/drive/My Drive/Colab Notebooks/DCEA_net_effect_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# SCREENING DETECTIONS

screening_plot = dcea_df[
    [
        "imd",
        "control_screen_detected",
        "intervention_screen_detected",
        "incremental_screen_detected"
    ]
].copy()

screening_plot = screening_plot.rename(columns={
    "imd": "IMD",
    "control_screen_detected": "NO_TEXT",
    "intervention_screen_detected": "SEND_TEXT",
    "incremental_screen_detected": "Incremental"
})

x = np.arange(len(screening_plot["IMD"]))
width = 0.25

fig, ax = plt.subplots(figsize=(12, 6))

ax.bar(
    x - width,
    screening_plot["NO_TEXT"],
    width,
    label="Before intervention"
)

ax.bar(
    x,
    screening_plot["SEND_TEXT"],
    width,
    label="After intervention"
)

ax.bar(
    x + width,
    screening_plot["Incremental"],
    width,
    label="Incremental screening detections"
)

ax.axhline(0, linewidth=0.8)

ax.set_xlabel("IMD decile")
ax.set_ylabel("Screening detections")
ax.set_title("Distribution of screening detections by IMD decile")

ax.set_xticks(x)
ax.set_xticklabels(screening_plot["IMD"])
ax.legend()

plt.tight_layout()

fig.savefig(
    "/content/drive/My Drive/Colab Notebooks/screening_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# SYMPTOMATIC DETECTIONS

symptomatic_plot = dcea_df[
    [
        "imd",
        "control_symptomatic",
        "intervention_symptomatic",
        "incremental_symptomatic"
    ]
].copy()

symptomatic_plot = symptomatic_plot.rename(columns={
    "imd": "IMD",
    "control_symptomatic": "NO_TEXT",
    "intervention_symptomatic": "SEND_TEXT",
    "incremental_symptomatic": "Incremental"
})

x = np.arange(len(symptomatic_plot["IMD"]))
width = 0.25

fig, ax = plt.subplots(figsize=(12, 6))

ax.bar(
    x - width,
    symptomatic_plot["NO_TEXT"],
    width,
    label="Before intervention"
)

ax.bar(
    x,
    symptomatic_plot["SEND_TEXT"],
    width,
    label="After intervention"
)

ax.bar(
    x + width,
    symptomatic_plot["Incremental"],
    width,
    label="Incremental symptomatic detections"
)

ax.axhline(0, linewidth=0.8)

ax.set_xlabel("IMD decile")
ax.set_ylabel("Symptomatic detections")
ax.set_title("Distribution of symptomatic detections by IMD decile")

ax.set_xticks(x)
ax.set_xticklabels(symptomatic_plot["IMD"])
ax.legend()

plt.tight_layout()

fig.savefig(
    "/content/drive/My Drive/Colab Notebooks/Symptomatic_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# 8.4 INEQUALITY MEASURES
# ----------------------------------

import statsmodels.api as sm

def calculate_inequality_measures(
    df,
    imd_col="imd",
    population_col="population",
    outcome_col="health_effect",
    analysis_type="single",
    control_col="control_health_effect",
    intervention_col="intervention_health_effect"
):
    """
    Calculate inequality measures for health effects or costs.

    analysis_type:
        "single"       -> Analyse one outcome column.
        "control"      -> Analyse the control arm.
        "intervention" -> Analyse the intervention arm.
        "comparison"   -> Analyse intervention minus control.

    The function can be used for both health effects and costs.
    For costs, specify the relevant cost column names.
    """

    data = df.copy()

    # Outcome to analyse

    if analysis_type == "single":
        selected_outcome = outcome_col
        data["analysis_outcome"] = data[selected_outcome]

    elif analysis_type == "control":
        selected_outcome = control_col
        data["analysis_outcome"] = data[selected_outcome]

    elif analysis_type == "intervention":
        selected_outcome = intervention_col
        data["analysis_outcome"] = data[selected_outcome]

    elif analysis_type == "comparison":
        data["analysis_outcome"] = (
            data[intervention_col] - data[control_col]
        )

    else:
        raise ValueError(
            "analysis_type must be 'single', 'control', "
            "'intervention', or 'comparison'."
        )

    outcome_col = "analysis_outcome"

    data = data[
        [imd_col, population_col, outcome_col]
    ].copy()

    data = data.sort_values(imd_col).reset_index(drop=True)

    if data[population_col].isna().any():
        raise ValueError("Population sizes cannot be missing.")

    if (data[population_col] <= 0).any():
        raise ValueError("Population sizes must be positive.")

    if data[outcome_col].isna().any():
        raise ValueError("Outcome values cannot be missing.")

    if data[imd_col].duplicated().any():
        raise ValueError("Each IMD decile must appear only once.")

    # Absolute and Relative Gap

    imd1_values = data.loc[
        data[imd_col] == 1, outcome_col
    ]
    imd10_values = data.loc[
        data[imd_col] == 10, outcome_col
    ]

    if imd1_values.empty or imd10_values.empty:
        raise ValueError("Data must include IMD deciles 1 and 10.")

    imd1 = imd1_values.iloc[0]
    imd10 = imd10_values.iloc[0]

    absolute_gap = imd10 - imd1

    relative_gap = (
        imd10 / imd1 if imd1 != 0 else np.nan
    )

    # Population-weighted relative rank

    data["population_share"] = (
        data[population_col] / data[population_col].sum()
    )

    # 0 = most deprived; 1 = least deprived
    data["rank"] = (
        data["population_share"].cumsum()
        - 0.5 * data["population_share"]
    )

    # SII: population-weighted regression

    X = sm.add_constant(data["rank"])

    model = sm.WLS(
        data[outcome_col],
        X,
        weights=data[population_col]
    ).fit()

    predicted_most_deprived = model.predict(
        pd.DataFrame({"const": [1], "rank": [0]})
    ).iloc[0]

    predicted_least_deprived = model.predict(
        pd.DataFrame({"const": [1], "rank": [1]})
    ).iloc[0]

    sii = (
        predicted_least_deprived
        - predicted_most_deprived
    )

    # RII

    rii = (
        predicted_least_deprived / predicted_most_deprived
        if not np.isclose(predicted_most_deprived, 0)
        else np.nan
    )

    # Concentration Index

    y = data[outcome_col].to_numpy()
    r = data["rank"].to_numpy()
    p = data["population_share"].to_numpy()

    mean_y = np.sum(p * y)
    mean_rank = np.sum(p * r)

    covariance = np.sum(
        p * (y - mean_y) * (r - mean_rank)
    )

    ci = (
        2 * covariance / mean_y
        if not np.isclose(mean_y, 0)
        else np.nan
    )

    # Results

    results = {
        "analysis_type": analysis_type,
        "absolute_gap": absolute_gap,
        "relative_gap": relative_gap,
        "SII": sii,
        "RII": rii,
        "CI": ci,
        "predicted_IMD1_end": predicted_most_deprived,
        "predicted_IMD10_end": predicted_least_deprived,
        "ranked_data": data,
        "regression": model
    }

    return results


# Cost inequality: intervention arm
control_cost_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="single",
    outcome_col="control_cost_per_woman"
)

print("\n=== CONTROL COST INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {control_cost_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {control_cost_inequality['relative_gap']:.2f}")
print(f"SII: {control_cost_inequality['SII']:.2f}")
print(f"RII: {control_cost_inequality['RII']:.2f}")
print(f"CI: {control_cost_inequality['CI']:.2f}")

# Cost inequality: control arm
intervention_cost_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="single",
    outcome_col="intervention_cost_per_woman"
)

print("\n=== INTERVENTION COST INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {intervention_cost_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {intervention_cost_inequality['relative_gap']:.2f}")
print(f"SII: {intervention_cost_inequality['SII']:.2f}")
print(f"RII: {intervention_cost_inequality['RII']:.2f}")
print(f"CI: {intervention_cost_inequality['CI']:.2f}")

# Incremental cost inequality
incremental_cost_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="comparison",
    control_col="control_cost_per_woman",
    intervention_col="intervention_cost_per_woman"
)
print("\n=== INCREMENTAL COST INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {incremental_cost_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {incremental_cost_inequality['relative_gap']:.2f}")
print(f"SII: {incremental_cost_inequality['SII']:.2f}")
print(f"RII: {incremental_cost_inequality['RII']:.2f}")
print(f"CI: {incremental_cost_inequality['CI']:.2f}")



# Net effect inequality: Control arm
control_net_effect_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="single",
    outcome_col="control_net_effect"
)
print("\n=== CONTROL NET EFFECT INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {control_net_effect_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {control_net_effect_inequality['relative_gap']:.2f}")
print(f"SII: {control_net_effect_inequality['SII']:.2f}")
print(f"RII: {control_net_effect_inequality['RII']:.2f}")
print(f"CI: {control_net_effect_inequality['CI']:.2f}")

# Net effect inequality: Intervention arm
intervention_net_effect_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="single",
    outcome_col="intervention_net_effect"
)
print("\n=== INTERVENTION NET EFFECT INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {intervention_net_effect_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {intervention_net_effect_inequality['relative_gap']:.2f}")
print(f"SII: {intervention_net_effect_inequality['SII']:.2f}")
print(f"RII: {intervention_net_effect_inequality['RII']:.2f}")
print(f"CI: {intervention_net_effect_inequality['CI']:.2f}")

# Inremental net effect inequality

incremental_net_effect_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="single",
    outcome_col="incremental_net_effect"
)

print("\n=== INCREMENTAL NET EFFECT INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {incremental_net_effect_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {incremental_net_effect_inequality['relative_gap']:.2f}")
print(f"SII: {incremental_net_effect_inequality['SII']:.2f}")
print(f"RII: {incremental_net_effect_inequality['RII']:.2f}")
print(f"CI: {incremental_net_effect_inequality['CI']:.2f}")

# Symptomatic detection inequality: Control arm
control_symptomatic_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="single",
    outcome_col="control_symptomatic"
)
print("\n=== CONTROL SYMPTOMATIC INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {control_symptomatic_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {control_symptomatic_inequality['relative_gap']:.2f}")
print(f"SII: {control_symptomatic_inequality['SII']:.2f}")
print(f"RII: {control_symptomatic_inequality['RII']:.2f}")
print(f"CI: {control_symptomatic_inequality['CI']:.2f}")

# Symptomatic detection inequality: Intervention arm
intervention_symptomatic_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="single",
    outcome_col="intervention_symptomatic"
)
print("\n=== INTERVENTION SYMPTOMATIC INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {intervention_symptomatic_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {intervention_symptomatic_inequality['relative_gap']:.2f}")
print(f"SII: {intervention_symptomatic_inequality['SII']:.2f}")
print(f"RII: {intervention_symptomatic_inequality['RII']:.2f}")
print(f"CI: {intervention_symptomatic_inequality['CI']:.2f}")

# Incremental symptomatic detection ineqaulity
incremental_symptomatic_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="single",
    outcome_col="incremental_symptomatic"
)
print("\n=== INCREMENTAL SYMPTOMATIC INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {incremental_symptomatic_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {incremental_symptomatic_inequality['relative_gap']:.2f}")
print(f"SII: {incremental_symptomatic_inequality['SII']:.2f}")
print(f"RII: {incremental_symptomatic_inequality['RII']:.2f}")
print(f"CI: {incremental_symptomatic_inequality['CI']:.2f}")

# Screening detection inequality: Control arm
control_screening_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="single",
    outcome_col="control_screen_detected"
)
print("\n=== CONTROL SCREENING INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {control_screening_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {control_screening_inequality['relative_gap']:.2f}")
print(f"SII: {control_screening_inequality['SII']:.2f}")
print(f"RII: {control_screening_inequality['RII']:.2f}")
print(f"CI: {control_screening_inequality['CI']:.2f}")

# Screening detection inequality: Intervention arm
intervention_screening_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="single",
    outcome_col="intervention_screen_detected"
)
print("\n=== INTERVENTION SCREENING INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {intervention_screening_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {intervention_screening_inequality['relative_gap']:.2f}")
print(f"SII: {intervention_screening_inequality['SII']:.2f}")
print(f"RII: {intervention_screening_inequality['RII']:.2f}")
print(f"CI: {intervention_screening_inequality['CI']:.2f}")

# Incremental screening detection inequality detection
incremental_screening_inequality = calculate_inequality_measures(
    dcea_df,
    analysis_type="single",
    outcome_col="incremental_screen_detected"
)

print("\n=== INCREMENTAL SCREENING INEQUALITY ===")
print(f"Absolute gap (IMD 10 - IMD 1): {incremental_screening_inequality['absolute_gap']:.2f}")
print(f"Relative gap (IMD 10 / IMD 1): {incremental_screening_inequality['relative_gap']:.2f}")
print(f"SII: {incremental_screening_inequality['SII']:.2f}")
print(f"RII: {incremental_screening_inequality['RII']:.2f}")
print(f"CI: {incremental_screening_inequality['CI']:.2f}")


# 8.5 EQUITY-WEIGHTED ANALYSIS
# ---------------------------------------

# Relative equity weights informed by the
# estimated 6.95:1 poorest-to-richest ratio.
MOST_DEPRIVED_WEIGHT = 6.95
LEAST_DEPRIVED_WEIGHT = 1.00

IMD_COL = "imd"
POPULATION_COL = "population"

outcome_cols = [
    "incremental_net_effect",
    "incremental_screen_detected",
    "incremental_symptomatic",
]


# 8.5.1 CALCULATE AND NORMALISE EQUITY WEIGHTS
# --------------------------------------------

def calculate_equity_weights(
    dcea_df,
    imd_col=IMD_COL,
    population_col=POPULATION_COL,
    most_deprived_weight=MOST_DEPRIVED_WEIGHT,
    least_deprived_weight=LEAST_DEPRIVED_WEIGHT,
):
    """
    Calculate deprivation-based equity weights for IMD deciles.

    Relative weights are linearly interpolated from 6.95 for
    IMD 1 (most deprived) to 1.00 for IMD 10 (least deprived).

    The weights are then normalised so that their
    population-weighted mean equals 1.

    Returns one row per IMD decile.
    """

    data = dcea_df[
        [imd_col, population_col]
    ].copy()

    # Check required columns
    if data[[imd_col, population_col]].isna().any().any():
        raise ValueError(
            "IMD and population columns cannot contain missing values."
        )

    data[imd_col] = pd.to_numeric(
        data[imd_col], errors="raise"
    )

    data[population_col] = pd.to_numeric(
        data[population_col], errors="raise"
    )

    # Check one row per decile
    if data[imd_col].duplicated().any():
        raise ValueError(
            "Expected exactly one row per IMD decile."
        )

    # Check population values
    if (
        ~np.isfinite(data[population_col])
    ).any() or (data[population_col] <= 0).any():
        raise ValueError(
            "Population values must be finite and positive."
        )

    # Check the ten IMD deciles
    expected_deciles = set(range(1, 11))
    observed_deciles = set(data[imd_col].tolist())

    if observed_deciles != expected_deciles:
        raise ValueError(
            "Expected IMD deciles 1 to 10, with IMD 1 "
            "being most deprived and IMD 10 least deprived."
        )

    if (
        not np.isfinite(most_deprived_weight)
        or not np.isfinite(least_deprived_weight)
        or most_deprived_weight <= 0
        or least_deprived_weight <= 0
        or most_deprived_weight < least_deprived_weight
    ):
        raise ValueError(
            "Equity weights must be finite, positive, and "
            "the most-deprived weight must be at least as large "
            "as the least-deprived weight."
        )

    # Sort deciles in deprivation order
    data = data.sort_values(imd_col).reset_index(drop=True)

    # Population shares
    data["population_share"] = (
        data[population_col] / data[population_col].sum()
    )

    # Linearly interpolate relative weights:
    # IMD 1 = 6.95; IMD 10 = 1.00
    data["relative_equity_weight"] = np.linspace(
        most_deprived_weight,
        least_deprived_weight,
        10,
    )

    # Population-weighted mean of relative weights
    normaliser = (
        data["population_share"]
        * data["relative_equity_weight"]
    ).sum()

    # Normalise weights so population-weighted mean = 1
    data["equity_weight"] = (
        data["relative_equity_weight"] / normaliser
    )

    # Verify normalisation
    weighted_mean = (
        data["population_share"]
        * data["equity_weight"]
    ).sum()

    if not np.isclose(weighted_mean, 1.0):
        raise RuntimeError(
            "Equity weights failed the normalisation check."
        )

    # Verify the relative weighting is preserved
    weight_ratio = (
        data.loc[
            data[imd_col] == 1, "equity_weight"
        ].iloc[0]
        /
        data.loc[
            data[imd_col] == 10, "equity_weight"
        ].iloc[0]
    )

    if not np.isclose(
        weight_ratio,
        most_deprived_weight / least_deprived_weight,
    ):
        raise RuntimeError(
            "Normalisation did not preserve the relative weight ratio."
        )

    return data[
        [
            imd_col,
            population_col,
            "population_share",
            "relative_equity_weight",
            "equity_weight",
        ]
    ]


# Calculate equity weights
equity_weights = calculate_equity_weights(dcea_df)


# Display weights
print("\nEQUITY WEIGHTS BY IMD DECILE")

print(
    equity_weights.round(4).to_string(index=False)
)

print(
    "\nPopulation-weighted mean of normalised weights:",
    round(
        (
            equity_weights["population_share"]
            * equity_weights["equity_weight"]
        ).sum(),
        6,
    ),
)

print(
    "Relative weight ratio (IMD 1 / IMD 10):",
    round(
        equity_weights.loc[
            equity_weights["imd"] == 1, "equity_weight"
        ].iloc[0]
        /
        equity_weights.loc[
            equity_weights["imd"] == 10, "equity_weight"
        ].iloc[0],
        6,
    ),
)



# 8.5.2 APPLY EQUITY WEIGHTS TO INCREMENTAL NET HEALTH EFFECT
# ------------------------------------------------------------

def calculate_weighted_health_benefit(
    dcea_df,
    equity_weights,
    imd_col="imd",
    population_col="population",
    outcome_col="incremental_net_effect",
):
    """
    Calculate total incremental net health benefit before and
    after applying normalised deprivation-based equity weights.

    The outcome is expressed per 100,000 people within each IMD
    decile. Population totals are calculated before aggregation.

    Returns:
        1. Decile-specific results.
        2. Aggregate unweighted and weighted health benefits.
    """

    data = dcea_df[
        [imd_col, population_col, outcome_col]
    ].copy()

    # Merge normalised equity weights
    data = data.merge(
        equity_weights[
            [
                imd_col,
                "population_share",
                "relative_equity_weight",
                "equity_weight",
            ]
        ],
        on=imd_col,
        how="left",
        validate="one_to_one",
    )

    if data["equity_weight"].isna().any():
        raise ValueError("Missing equity weights for one or more IMD deciles.")

    # Incremental net health effect before weighting
    data["unweighted_incremental_net_effect"] = data[outcome_col]

    # Equity-weighted incremental net health benefit
    data["weighted_incremental_net_effect"] = (
        data["unweighted_incremental_net_effect"]
        * data["equity_weight"]
    )

    # Difference attributable to equity weighting
    data["weighting_difference"] = (
        data["weighted_incremental_net_effect"]
        - data["unweighted_incremental_net_effect"]
    )

    # Aggregate total health benefits across all deciles
    total_unweighted = data["unweighted_incremental_net_effect"].sum()
    total_weighted = data["weighted_incremental_net_effect"].sum()

    aggregate_results = pd.DataFrame({
        "measure": [
            "Total incremental net effect",
            "Total equity-weighted incremental net effect benefit",
            "Difference after applying equity weights",
        ],
        "health_benefit": [
            total_unweighted,
            total_weighted,
            total_weighted - total_unweighted,
        ],
    })

    decile_results = data[
        [
            imd_col,
            population_col,
            outcome_col,
            "relative_equity_weight",
            "equity_weight",
            "unweighted_incremental_net_effect",
            "weighted_incremental_net_effect",
            "weighting_difference",
        ]
    ].sort_values(imd_col).reset_index(drop=True)

    return decile_results, aggregate_results

# RUN ANALYSIS

decile_health_benefit, aggregate_health_benefit = (
    calculate_weighted_health_benefit(
        dcea_df=dcea_df,
        equity_weights=equity_weights,
    )
)

# RESULTS

print("\nTOTAL INCREMENTAL NET HEALTH BENEFIT BY IMD DECILE")

print(
    decile_health_benefit.round(4).to_string(index=False)
)

print("\nAGGREGATE HEALTH BENEFIT BEFORE AND AFTER WEIGHTING")

print(
    aggregate_health_benefit.round(4).to_string(index=False)
)

# 8.5.3 PLOT OF EQUITY WEIGHTING
# ---------------------------------

import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator, StrMethodFormatter
import numpy as np

# Extract results
plot_data = decile_health_benefit.sort_values("imd")

imd = plot_data["imd"].astype(int)

before_weighting = plot_data[
    "unweighted_incremental_net_effect"
]

after_weighting = plot_data[
    "weighted_incremental_net_effect"
]

# Set bar positions
x = np.arange(len(imd))
width = 0.38

# Create figure
fig, ax = plt.subplots(figsize=(12, 6))

# Plot grouped bars
ax.bar(
    x - width/2,
    before_weighting,
    width,
    label="Before equity weighting"
)

ax.bar(
    x + width/2,
    after_weighting,
    width,
    label="After equity weighting"
)

# Labels and formatting
ax.set_title(
    "Incremental net effect before and after equity weighting",
    fontsize=15
)

ax.set_xlabel(
    "IMD decile (1 = most deprived, 10 = least deprived)",
    fontsize=12
)

ax.set_ylabel(
    "Total incremental net effect",
    fontsize=12
)

ax.set_xticks(x)
ax.set_xticklabels(imd)

# Ensure whole numbers on y-axis
ax.yaxis.set_major_locator(MaxNLocator(integer=True))
ax.yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))

# Start y-axis at zero
ax.set_ylim(bottom=0)

# Legend and grid
ax.legend(fontsize=11)
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()

fig.savefig(
    "/content/drive/My Drive/Colab Notebooks/weighted_analysis.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()




```
